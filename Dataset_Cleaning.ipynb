```python
keys = ['PATNO', 'EVENT_ID']

# ── p4, p5, p6 — select only what staging needs ───────────────
p4 = p4_clean[keys + [c for c in ['NP1COG','NP1RTOT']
                       if c in p4_clean.columns]].copy()

p5 = p5_clean[keys + [c for c in ['NP1PTOT']
                       if c in p5_clean.columns]].copy()

p6 = p6_clean[keys + [c for c in ['NP2PTOT']
                       if c in p6_clean.columns]].copy()

# ── Merge — df3_f is already clean, use directly ──────────────
updrs = (df3_new_clean
         .merge(p4, on=keys, how='left')
         .merge(p5, on=keys, how='left')
         .merge(p6, on=keys, how='left'))

# ── Rename to staging names ───────────────────────────────────
updrs = updrs.rename(columns={
    'NP1COG':  'COG1',
    'NP1RTOT': 'Part1',
    'NP1PTOT': 'Part1P',
    'NP2PTOT': 'Part2',
    'NP3TOT':  'Part3',
    'NHY':     'HY_STAGE_NUM'
})

# ── Derived columns ───────────────────────────────────────────
updrs['Part1_excl_COG1'] = updrs['Part1'] - updrs['COG1']

updrs['subthreshold_parkinsonism'] = updrs['Part3'].apply(
    lambda x: 1 if pd.notna(x) and x >= 5
              else (np.nan if pd.isna(x) else 0)
)

# ── Verify ────────────────────────────────────────────────────
print(f"Shape:           {updrs.shape}")
print(f"Unique patients: {updrs['PATNO'].nunique()}")
print(f"Dupes:           {updrs.duplicated(subset=keys).sum()}")

print(f"\nNull counts:")
for col in ['COG1','Part1','Part1P','Part2','Part3','HY_STAGE_NUM']:
    if col in updrs.columns:
        n   = updrs[col].isna().sum()
        pct = n / len(updrs) * 100
        flag = "✅" if pct < 20 else ("⚠️" if pct < 50 else "❌")
        print(f"  {flag} {col}: {n} ({pct:.1f}%)")
```

```python
# Base: status + UPDRS (already merged)
final_df = status_updrs2.copy()

# Define visit keys once
keys = ["PATNO", "EVENT_ID"]

for name, df, key in [
    ("MoCA",         df10_clean, keys),
    ("Symbol Digit", df11_clean, keys),
    # ("Diagnosis",    df2_clean,  ["PATNO"]),
    ("Demographics", df1_clean,  ["PATNO"]),
]:
    before = len(final_df)

    merged = final_df.merge(
        df,
        on=key,
        how="left",
        suffixes=("", "_dup")
    )

    # Drop duplicate columns coming from the right
    dup_cols = [c for c in merged.columns if c.endswith("_dup")]
    merged.drop(columns=dup_cols, inplace=True)

    after = len(merged)

    # For dup check, use PATNO+EVENT_ID when available
    if set(keys).issubset(merged.columns):
        dupes = merged.duplicated(subset=keys).sum()
    else:
        dupes = merged.duplicated(subset=["PATNO"]).sum()

    status_flag = "✅" if after == before and dupes == 0 else "❌ STOP"
    print(f"{name:<15} {before} → {after} rows  dupes={dupes}  {status_flag}")

    # Update final_df for the next iteration
    final_df = merged

# ── Final check ───────────────────────────────────────────────
print(f"\nFinal shape:        {final_df.shape}")
print(f"Unique patients:    {final_df['PATNO'].nunique()}")
print(f"Dupes (PATNO+EVENT): {final_df.duplicated(subset=keys).sum()}")
```

```python
import pandas as pd
import numpy as np

# ── Load ──────────────────────────────────────────────
# saa = pd.read_csv('SAA_Biospecimen_Analysis_Results.csv')

# ── Step 1: Priority order for SAA_Status ─────────────
# Positive > Inconclusive > Negative
# Biologically: if ever positive → patient is S+
status_priority = {
    'Positive':     3,
    'Inconclusive': 2,
    'Negative':     1,
}
saa['status_rank'] = saa['SAA_Status']\
                        .map(status_priority)\
                        .fillna(0)

# ── Step 2: Priority order for PROJECTID ──────────────
# 237 (Amprion 24h) > 155 (older method)
saa['project_rank'] = saa['PROJECTID']\
                         .map({237: 2, 155: 1})\
                         .fillna(0)

# ── Step 3: Sort → take most severe + best project ────
saa_sorted = saa.sort_values(
    ['PATNO', 'status_rank', 'project_rank'],
    ascending=[True, False, False]
)

# Keep highest status per patient
saa_dedup = saa_sorted.drop_duplicates(
    'PATNO', keep='first').copy()

# ── Step 4: Binary S anchor ───────────────────────────
saa_dedup['SAA_pos'] = (
    saa_dedup['SAA_Status'] == 'Positive'
).astype(int)

# Inconclusive → conservative → treat as 0
# but flag separately
saa_dedup['SAA_inconclusive'] = (
    saa_dedup['SAA_Status'] == 'Inconclusive'
).astype(int)

# ── Step 5: SAA_Type encoded ──────────────────────────
# Type1=2, Type2=1, NaN/NA=0
saa_dedup['SAA_Type_encoded'] = saa_dedup['SAA_Type']\
    .map({'Type1': 2, 'Type2': 1})\
    .fillna(0).astype(int)

# ── Step 6: Final columns ─────────────────────────────
saa_final = saa_dedup[[
    'PATNO',
    'EVENT_ID',
    'SAA_Status',          # for staging logic
    'SAA_pos',             # S anchor binary
    'SAA_inconclusive',    # flag for QC
    'SAA_Type_encoded',    # feature for model
]].reset_index(drop=True)

# ── Verify ────────────────────────────────────────────
print("=== SAA FINAL ===")
print(f"Shape:         {saa_final.shape}")
print(f"Unique PATNO:  {saa_final['PATNO'].nunique()}")
print(f"Dupes:         {saa_final.duplicated('PATNO').sum()}")
print(f"\nSAA_pos dist:")
print(saa_final['SAA_pos'].value_counts(dropna=False))
print(f"\nSAA_inconclusive: {saa_final['SAA_inconclusive'].sum()}")
print(f"SAA_Type_encoded dist:")
print(saa_final['SAA_Type_encoded'].value_counts(dropna=False))
```

```python
# rbd = pd.read_csv('REM_Sleep_Disorder_Questionnaire.csv')

# ── RBD items (12 items, not medical history) ─────────
RBD_ITEMS = [
    'DRMVIVID',  # vivid dreams
    'DRMAGRAC',  # aggressive dreams
    'DRMNOCTB',  # nocturnal behavior
    'SLPLMBMV',  # limb movements
    'SLPINJUR',  # sleep injury
    'DRMVERBL',  # verbal behavior
    'DRMFIGHT',  # fighting
    'DRMUMV',    # unusual movements
    'DRMOBJFL',  # object falls
    'MVAWAKEN',  # movements awaken
    'DRMREMEM',  # dream recall
    'SLPDSTRB',  # sleep disturbance
]
# IGNORE: STROKE, HETRA, PARKISM, RLS,
# NARCLPSY, DEPRS, EPILEPSY, BRNINFM, CNSOTH
# These are comorbidity history — NOT RBD items

# ── Step 1: Convert items to numeric ──────────────────
for col in RBD_ITEMS:
    rbd[col] = pd.to_numeric(
        rbd[col], errors='coerce')

# ── Step 2: Compute RBDTOT per row ────────────────────
# Sum of 12 items (0-12 range)
# skipna=True handles the few nulls
rbd['RBDTOT'] = rbd[RBD_ITEMS]\
                    .sum(axis=1, skipna=True)

# ── Step 3: RBD positive flag per row ─────────────────
rbd['RBD_pos'] = (rbd['RBDTOT'] >= 5).astype(int)

# ── Step 4: Aggregate to patient level ────────────────
# Strategy: take MAX across all visits
# If ever RBD positive → patient is RBD+
# Take MAX RBDTOT (most severe recorded)
rbd_agg = rbd.groupby('PATNO').agg(
    RBDTOT  = ('RBDTOT',  'max'),
    RBD_pos = ('RBD_pos', 'max'),
    rbd_n_visits = ('RBDTOT', 'count')
).reset_index()

# ── Step 5: Final columns ─────────────────────────────
rbd_final = rbd_agg[[
    'PATNO',
    'RBDTOT',        # continuous score (for model)
    'RBD_pos',       # binary flag (for staging)
    'rbd_n_visits',  # QC: how many visits
]].copy()

# ── Verify ────────────────────────────────────────────
print("=== RBD FINAL ===")
print(f"Shape:        {rbd_final.shape}")
print(f"Unique PATNO: {rbd_final['PATNO'].nunique()}")
print(f"Dupes:        {rbd_final.duplicated('PATNO').sum()}")
print(f"\nRBDTOT stats:")
print(rbd_final['RBDTOT'].describe())
print(f"\nRBD_pos dist:")
print(rbd_final['RBD_pos'].value_counts(dropna=False))
print(f"RBD positive (>=5): {rbd_final['RBD_pos'].sum()}")
print(f"Missing RBDTOT:     {rbd_final['RBDTOT'].isna().sum()}")
```

```python
# upsit = pd.read_csv('University_of_Pennsylvania_Smell_ID_Test.csv')

# ── Step 1: Compute TOTAL if missing ──────────────────
# TOTAL_CORRECT = sum of SCENT_XX_CORRECT columns
scent_cols = [f'SCENT_{i:02d}_CORRECT'
              for i in range(1, 41)]
scent_avail = [c for c in scent_cols
               if c in upsit.columns]

upsit['TOTAL_CORRECT'] = pd.to_numeric(
    upsit['TOTAL_CORRECT'], errors='coerce')

# Fill missing totals from items
null_mask = upsit['TOTAL_CORRECT'].isna()
if null_mask.any():
    upsit.loc[null_mask, 'TOTAL_CORRECT'] = \
        upsit.loc[null_mask, scent_avail]\
             .apply(pd.to_numeric, errors='coerce')\
             .sum(axis=1)

# ── Step 2: Aggregate to patient level ────────────────
# Strategy: take MIN across visits
# Lowest UPSIT = worst smell = most severe
# This is most conservative for staging
upsit_agg = upsit.groupby('PATNO').agg(
    UPSITTOTAL   = ('TOTAL_CORRECT', 'min'),
    UPSIT_mean   = ('TOTAL_CORRECT', 'mean'),
    upsit_n_visits = ('TOTAL_CORRECT', 'count')
).reset_index()

# ── Step 3: Hyposmia flag ─────────────────────────────
# < 25/40 = hyposmic (validated threshold)
upsit_agg['hyposmic'] = (
    upsit_agg['UPSITTOTAL'] < 25
).astype(int)

# ── Step 4: Final columns ─────────────────────────────
upsit_final = upsit_agg[[
    'PATNO',
    'UPSITTOTAL',      # min score (for staging)
    'UPSIT_mean',      # mean score (for model)
    'hyposmic',        # binary flag (for staging)
    'upsit_n_visits',  # QC
]].copy()

# ── Verify ────────────────────────────────────────────
print("=== UPSIT FINAL ===")
print(f"Shape:        {upsit_final.shape}")
print(f"Unique PATNO: {upsit_final['PATNO'].nunique()}")
print(f"Dupes:        {upsit_final.duplicated('PATNO').sum()}")
print(f"\nUPSITTOTAL stats:")
print(upsit_final['UPSITTOTAL'].describe())
print(f"\nHyposmic (<25): {upsit_final['hyposmic'].sum()}")
print(f"Normal  (>=25): {(upsit_final['hyposmic']==0).sum()}")
print(f"Missing:        {upsit_final['UPSITTOTAL'].isna().sum()}")
```

```python
import numpy as np
import pandas as pd

def add_biospecimen_to_final(final_df,
                              saa_final,
                              rbd_final,
                              upsit_final,
                              dat_final,
                              med_final):
    df = final_df.copy()

    hc_mask = (df['COHORT_DEFINITION']
               == 'Healthy Control')
    pd_mask = (df['COHORT_DEFINITION']
               == "Parkinson's Disease")

    def safe_merge(base_df, target_df):
        """
        Merge on PATNO+EVENT_ID if both have it.
        Otherwise merge on PATNO only.
        No sorting — all targets already deduplicated.
        """
        if ('EVENT_ID' in base_df.columns and
            'EVENT_ID' in target_df.columns):
            return base_df.merge(
                target_df,
                on=['PATNO','EVENT_ID'],
                how='left')

        target_clean = target_df\
            .drop_duplicates(subset='PATNO',
                             keep='first')
        return base_df.merge(
            target_clean, on='PATNO', how='left')

    # ── SAA ───────────────────────────────────────
    df = safe_merge(df, saa_final)
    df.loc[hc_mask, 'SAA_pos']          = 0
    df.loc[hc_mask, 'SAA_Type_encoded'] = 0
    df.loc[hc_mask, 'SAA_inconclusive'] = 0
    pd_no_saa = pd_mask & df['SAA_pos'].isna()
    df.loc[pd_no_saa, 'SAA_pos']          = 1
    df.loc[pd_no_saa, 'SAA_Type_encoded'] = 0
    df.loc[pd_no_saa, 'SAA_inconclusive'] = 0

    # ── DATScan ───────────────────────────────────
    df = safe_merge(df, dat_final)
    df.loc[hc_mask, 'DAT_pos'] = 0
    pd_no_dat = pd_mask & df['DAT_pos'].isna()
    df.loc[pd_no_dat, 'DAT_pos'] = 1

    # ── RBD ───────────────────────────────────────
    df = safe_merge(df, rbd_final)

    # Impute RBDTOT only — do NOT recompute RBD_pos
    if 'EVENT_ID' in df.columns:
        rbd_ev_med = df.groupby(
            ['COHORT_DEFINITION','EVENT_ID']
        )['RBDTOT'].transform('median')
        df['RBDTOT'] = df['RBDTOT'].fillna(rbd_ev_med)

    rbd_coh_med = df.groupby(
        'COHORT_DEFINITION')['RBDTOT']\
        .transform('median')
    df['RBDTOT'] = df['RBDTOT'].fillna(rbd_coh_med)

    # Keep RBD_pos from original data → just fill NaN=0
    df['RBD_pos'] = df['RBD_pos']\
                      .fillna(0).astype(int)

    # ── UPSIT ─────────────────────────────────────
    df = safe_merge(df, upsit_final)
    upsit_coh_med = df.groupby(
        'COHORT_DEFINITION')['UPSITTOTAL']\
        .transform('median')
    df['UPSITTOTAL'] = df['UPSITTOTAL']\
                         .fillna(upsit_coh_med)
    df['UPSIT_mean'] = df.get('UPSIT_mean',
                               df['UPSITTOTAL'])
    # hyposmic: only from real data
    # Imputed → use conservative (not hyposmic)
    df['hyposmic'] = df['hyposmic']\
                       .fillna(0).astype(int)

    # ── LEDD ──────────────────────────────────────
    df = safe_merge(df, med_final)
    df.loc[hc_mask, 'LEDD_total']   = 0
    df.loc[hc_mask, 'LEDD_max']     = 0
    df.loc[hc_mask, 'LEDD_on_meds'] = 0
    df['LEDD_total']   = df['LEDD_total'].fillna(0)
    df['LEDD_max']     = df['LEDD_max'].fillna(0)
    df['LEDD_on_meds'] = df['LEDD_on_meds']\
                           .fillna(0).astype(int)

    # ── PDMEDYN ───────────────────────────────────
    df.loc[hc_mask, 'PDMEDYN'] = 0
    df['PDMEDYN'] = df['PDMEDYN']\
                      .fillna(0).astype(int)

    # ── Coverage Report ───────────────────────────
    print(f"\n{'='*55}")
    print("FINAL COVERAGE REPORT")
    print(f"{'='*55}")
    print(f"Shape:        {df.shape}")
    print(f"Unique PATNO: {df['PATNO'].nunique()}")

    check = {
        'SAA_pos':    'S anchor (SAA)',
        'DAT_pos':    'D anchor (DATScan)',
        'PDMEDYN':    'PD Medications',
        'LEDD_total': 'LEDD Total',
        'RBDTOT':     'RBD Score',
        'RBD_pos':    'RBD Positive',
        'UPSITTOTAL': 'UPSIT Score',
        'hyposmic':   'Hyposmic Flag',
    }

    print(f"\n{'Feature':<25} {'Cov%':>6} "
          f"{'Null':>6} {'HC mean':>9} {'PD mean':>9}")
    print("-" * 58)

    for col, name in check.items():
        if col in df.columns:
            cov  = df[col].notna().mean() * 100
            null = df[col].isna().sum()
            hc_m = df.loc[hc_mask, col].mean()
            pd_m = df.loc[pd_mask, col].mean()
            print(f"  {name:<23} {cov:>5.1f}% "
                  f"{null:>6} {hc_m:>9.3f} "
                  f"{pd_m:>9.3f}")
        else:
            print(f"  {name:<23} MISSING")

    return df


# ── Run AFTER fixing PATNO types ──────────────────────
def fix_patno(df):
    df['PATNO'] = pd.to_numeric(
        df['PATNO'], errors='coerce')\
        .astype('Int64')
    return df

final_df    = fix_patno(final_df)
saa_final   = fix_patno(saa_final)
rbd_final   = fix_patno(rbd_final)
upsit_final = fix_patno(upsit_final)
dat_final   = fix_patno(dat_final)
med_final   = fix_patno(med_final)

# Verify overlap first
print("PATNO overlap check:")
for name, df_c in [('SAA',   saa_final),
                   ('RBD',   rbd_final),
                   ('UPSIT', upsit_final),
                   ('DAT',   dat_final),
                   ('MED',   med_final)]:
    ov = set(final_df['PATNO'].dropna()) & \
         set(df_c['PATNO'].dropna())
    print(f"  {name:<8}: {len(ov)} patients overlap")

# Run
final_df_new = add_biospecimen_to_final(
    final_df    = final_df,
    saa_final   = saa_final,
    rbd_final   = rbd_final,
    upsit_final = upsit_final,
    dat_final   = dat_final,
    med_final   = med_final
)
```

```python
import pandas as pd
import numpy as np

def assign_nsd_iss_stage(row):
    # ── Biologic Anchors ──────────────────────────────────────
    S = row['SAA_pos']
    D = row['DAT_pos']

    # ── Clinical Indicators ────────────────────────────────────
    item11   = row['COG1']
    moca     = row['MCATOT']
    sub_park = row['subthreshold_parkinsonism']
    on_meds  = row['PDMEDYN']
    rbd      = row['RBD_pos']
    hyposmic = row['hyposmic']
    updrs2   = row['Part2']
    updrs1   = row['Part1_excl_COG1']

    # ── Non-Biologic / Unstageable Filtering ───────────────────
    # Patient must be SAA positive to enter the MDS-ISS staging framework
    if pd.isna(S) or S == 0:
        return 'Non-PD / Control'

    # Crucial structural missing check
    if pd.isna(item11):
        return None

    # ── STAGE 6: Severe Functional Impairment ──────────────────
    if ((item11 == 4 and not pd.isna(moca) and moca <= 24) or
        (not pd.isna(updrs2) and updrs2 >= 40) or
        (not pd.isna(updrs1) and updrs1 >= 37)):
        return '6'

    # ── STAGE 5: Severe Clinical Signs / Symptoms ──────────────
    if ((item11 == 3 and not pd.isna(moca) and moca <= 24) or
        (item11 == 4 and not pd.isna(moca) and moca >= 25) or
        (not pd.isna(updrs2) and 27 <= updrs2 <= 39) or
        (not pd.isna(updrs1) and 25 <= updrs1 <= 36)):
        return '5'

    # ── STAGE 4: Moderate Clinical Signs / Symptoms ────────────
    if ((item11 == 2 and not pd.isna(moca) and moca <= 24) or
        (item11 == 3 and not pd.isna(moca) and moca >= 25) or
        (not pd.isna(updrs2) and 14 <= updrs2 <= 26) or
        (not pd.isna(updrs1) and 13 <= updrs1 <= 24)):
        return '4'

    # ── STAGE 3: Mild Functional Impairment ────────────────────
    stage3_cog = ((item11 == 1 and not pd.isna(moca) and moca <= 24) or
                  (item11 == 2 and not pd.isna(moca) and moca >= 25))

    stage3_mot = (not pd.isna(updrs2) and (3 <= updrs2 <= 13) and
                  (sub_park == 1 or on_meds == 1))

    if stage3_cog or stage3_mot:
        return '3'

    # ── STAGE 2: Mild Clinical Signs / Symptoms (A/B) ──────────
    stage2_cog   = (item11 == 1 and not pd.isna(moca) and moca >= 25)
    stage2_motor = (sub_park == 1 or on_meds == 1)
    stage2_nm    = (rbd == 1 or hyposmic == 1)

    if stage2_cog or stage2_motor or stage2_nm:
        return '2B' if D == 1 else '2A'

    # ── STAGE 1: No Clinical Signs / Symptoms (A/B) ────────────
    stage1_criteria = (
        item11 == 0 and
        sub_park == 0 and
        on_meds == 0 and
        rbd == 0 and
        hyposmic == 0
    )
    if stage1_criteria:
        return '1B' if D == 1 else '1A'

    # ── STAGE 0: Prodromal/Biologic Anchor Only ────────────────
    # S+ but all objective clinical and motor tests are totally negative (-)
    return '0'

# ── Execute ────────────────────────────────────────────────────
final_df_new['NSD_ISS_raw'] = final_df_new.apply(assign_nsd_iss_stage, axis=1)

print(final_df_new['NSD_ISS_raw'].value_counts(dropna=False).sort_index())
```

```python
def assign_nsd_iss_stage(row):
    """
    NSD-ISS staging for PD/Prodromal cohort only.
    HC cohort is handled separately — not touched here.
    """

    # ── Only stage PD and Prodromal ───────────────
    # HC rows never enter this function
    cohort = row.get('COHORT_DEFINITION', '')
    if cohort == 'Healthy Control':
        return 0   # HC = Stage 0 by definition

    # ── Biologic Anchors ──────────────────────────
    S = row.get('SAA_pos', np.nan)
    D = row.get('DAT_pos', np.nan)

    # ── Clinical Variables ────────────────────────
    item11 = row.get('COG1',             np.nan)
    moca   = row.get('MCATOT',           np.nan)
    updrs2 = row.get('Part2',            np.nan)
    updrs1 = row.get('Part1_excl_COG1',  np.nan)

    # Binary flags — NaN treated conservatively as 0
    sub_park = int(row.get('subthreshold_parkinsonism', 0) == 1)
    on_meds  = int(row.get('PDMEDYN',  0) == 1)
    rbd      = int(row.get('RBD_pos',  0) == 1)
    hyposmic = int(row.get('hyposmic', 0) == 1)

    # ── S- PD patient → Stage 0 ───────────────────
    # SAA negative PD exists (~15% of cases)
    # Biologically: no alpha-synuclein pathology detected
    if pd.isna(S) or S == 0:
        return 0

    # ── From here: S = 1 (SAA positive) ──────────
    # Need item11 to assign clinical stage
    if pd.isna(item11):
        return np.nan   # cannot classify without it

    D_pos = (not pd.isna(D) and D == 1)

    # ── STAGE 6: S+ D+ ────────────────────────────
    if D_pos:
        cog6 = (item11 == 4 and
                not pd.isna(moca) and moca <= 24)
        mot6 = (not pd.isna(updrs2) and updrs2 >= 40)
        nm6  = (not pd.isna(updrs1) and updrs1 >= 37)
        if cog6 or mot6 or nm6:
            return 6

    # ── STAGE 5: S+ D+ ────────────────────────────
    if D_pos:
        cog5 = ((item11 == 3 and
                 not pd.isna(moca) and moca <= 24) or
                (item11 == 4 and
                 not pd.isna(moca) and moca >= 25))
        mot5 = (not pd.isna(updrs2) and
                27 <= updrs2 <= 39)
        nm5  = (not pd.isna(updrs1) and
                25 <= updrs1 <= 36)
        if cog5 or mot5 or nm5:
            return 5

    # ── STAGE 4: S+ D+ ────────────────────────────
    if D_pos:
        cog4 = ((item11 == 2 and
                 not pd.isna(moca) and moca <= 24) or
                (item11 == 3 and
                 not pd.isna(moca) and moca >= 25))
        mot4 = (not pd.isna(updrs2) and
                14 <= updrs2 <= 26)
        nm4  = (not pd.isna(updrs1) and
                13 <= updrs1 <= 24)
        if cog4 or mot4 or nm4:
            return 4

    # ── STAGE 3: S+ D+ ────────────────────────────
    if D_pos:
        cog3 = ((item11 == 1 and
                 not pd.isna(moca) and moca <= 24) or
                (item11 == 2 and
                 not pd.isna(moca) and moca >= 25))
        mot3 = (not pd.isna(updrs2) and
                3 <= updrs2 <= 13 and
                (sub_park == 1 or on_meds == 1))
        if cog3 or mot3:
            return 3

    # ── STAGE 2: S+ D± ────────────────────────────
    # 2A = S+ D-    2B = S+ D+
    cog2 = (item11 == 1 and
             not pd.isna(moca) and moca >= 25)
    mot2 = (sub_park == 1 or on_meds == 1)
    nm2  = (rbd == 1 or hyposmic == 1)

    if cog2 or mot2 or nm2:
        return '2B' if D_pos else '2A'

    # ── STAGE 1: S+ D± ────────────────────────────
    # 1A = S+ D-    1B = S+ D+
    # ALL criteria must be absent
    crit1 = (item11 == 0 and
              sub_park == 0 and
              on_meds  == 0 and
              rbd      == 0 and
              hyposmic == 0)

    if crit1:
        return '1B' if D_pos else '1A'

    # ── Unclassifiable ────────────────────────────
    return np.nan


# ── Apply ─────────────────────────────────────────────
# ── Step 1: Apply staging ─────────────────────────────
final_df_new['NSD_ISS_raw'] = final_df_new.apply(
    assign_nsd_iss_stage, axis=1)

# ── Step 2: View distribution BEFORE dropping ─────────
print("=== RAW DISTRIBUTION ===")
print(final_df_new.groupby('COHORT_DEFINITION')[
    'NSD_ISS_raw']\
    .apply(lambda x:
           x.astype(str).value_counts())\
    .to_string())

# ── Step 3: Separate HC from PD ───────────────────────
hc_df = final_df_new[
    final_df_new['COHORT_DEFINITION']
    == 'Healthy Control'
].copy()

pd_df = final_df_new[
    final_df_new['COHORT_DEFINITION']
    != 'Healthy Control'
].copy()

print(f"\nHC rows:  {len(hc_df)}")
print(f"PD rows:  {len(pd_df)}")

# ── Step 4: Drop Stage 0 from PD only ─────────────────
# Stage 0 PD = SAA negative = cannot be staged
pd_stage0 = (pd_df['NSD_ISS_raw'] == 0)
pd_nan    = pd_df['NSD_ISS_raw'].isna()

print(f"\nPD Stage 0 (SAA-): {pd_stage0.sum()} → DROP")
print(f"PD Unclassifiable: {pd_nan.sum()} → DROP")

pd_staged = pd_df[
    ~pd_stage0 & ~pd_nan
].copy()

print(f"PD remaining after drop: {len(pd_staged)}")

# # ── Map to merged classes ─────────────────────────────
# stage_to_class = {
#     0:    0,   # Stage 0  → HC / No biomarkers
#     '1A': 1,   # Early
#     '1B': 1,   # Early
#     '2A': 1,   # Early
#     '2B': 1,   # Early
#     3:    2,   # Moderate
#     4:    2,   # Moderate
#     5:    3,   # Advanced
#     6:    3,   # Advanced
# }

# final_df_new['stage_merged'] = \
#     final_df_new['NSD_ISS_raw'].map(stage_to_class)

# # ── Final distribution ────────────────────────────────
# print("\nMerged class distribution:")
# label_names = {
#     0: 'HC / Stage 0',
#     1: 'Early (Stage 1-2)',
#     2: 'Moderate (Stage 3-4)',
#     3: 'Advanced (Stage 5-6)',
# }
# for cls, name in label_names.items():
#     n = (final_df_new['stage_merged'] == cls).sum()
#     print(f"  {name:<25}: {n}")

# print(f"\n  Unclassifiable (NaN): "
#       f"{final_df_new['stage_merged'].isna().sum()}")
# print(f"  Total rows:           {len(final_df_new)}")
```

```python
# ══════════════════════════════════════════════════════════════
# PART 2 — NSD-ISS STAGING FUNCTION PER PPMI GUIDE
# ══════════════════════════════════════════════════════════════

def assign_nsd_iss_ppmi(row):
    """
    NSD-ISS staging per PPMI operationalized criteria
    Dam et al. 2024, Lancet Neurology

    Biological anchors:
        S = SAA_Binary    (S+ / S- / S_unknown)
        D = D_pos         (D+ / D- / D_unknown)
        G = snca_variant  (1 = SNCA mutation)

    Clinical anchors:
        COG1       = MDS-UPDRS item 1.1 (0-4)
        MCATOT     = MoCA total (0-30)
        Part2      = MDS-UPDRS Part II
        Part1_excl = MDS-UPDRS Part I excl. item 1.1
        subthresh  = Part3 >= 5
        on_meds    = on PD medication
        rbd        = RBD confirmed
        hyposmic   = UPSIT <= 15th percentile
    """

    s    = row.get('S_status',  'S_unknown')
    d    = row.get('D_status',  'D_unknown')
    snca = row.get('snca_variant', 0)

    cog1     = row.get('COG1', None)
    moca     = row.get('MCATOT', None)
    part2    = row.get('Part2', None)
    part1_ex = row.get('Part1_excl_COG1', None)
    part3    = row.get('Part3', None)
    subthresh= row.get('subthreshold_parkinsonism', 0)
    on_meds  = row.get('on_pd_meds', 0)
    rbd      = row.get('rbd_flag', None)
    hyposmic = row.get('hyposmic', None)

    fallbacks    = []
    has_rbd      = (rbd == 1)
    has_hyposmic = (pd.notna(hyposmic) and hyposmic == 1)
    has_subthresh= (subthresh == 1)
    is_on_meds   = (on_meds == 1)

    # ── Fallback: infer S from clinical if SAA missing ────────
    if s == 'S_unknown':
        if d == 'D+' or has_subthresh or is_on_meds:
            s = 'S_inferred'
            fallbacks.append('S inferred from D+/motor signs')
        else:
            return 'Unstaged', 'SAA missing — no fallback'

    if s not in ('S+', 'S_inferred'):
        return 'Not NSD', ''

    # ── Fallback: infer D from Part3 if DAT missing ───────────
    if d == 'D_unknown':
        if pd.notna(part3) and part3 >= 5:
            d = 'D_inferred'
            fallbacks.append('D inferred from UPDRS-III >= 5')

    fb = '; '.join(fallbacks)

    # ──────────────────────────────────────────────────────────
    # STAGE 0: SNCA mutation, no pathological alpha-syn
    # ──────────────────────────────────────────────────────────
    if snca == 1 and s not in ('S+', 'S_inferred'):
        return 'Stage 0', fb

    # ──────────────────────────────────────────────────────────
    # STAGE 1A: S+ D-  COG1=0, no clinical signs
    # ──────────────────────────────────────────────────────────
    if d not in ('D+', 'D_inferred'):
        no_signs = (not has_subthresh and not is_on_meds
                    and not has_rbd and not has_hyposmic)
        if pd.notna(cog1) and cog1 == 0 and no_signs:
            return 'Stage 1A', fb

        # STAGE 2A: S+ D-, subtle signs present
        cog_2a = (pd.notna(cog1) and pd.notna(moca)
                  and cog1 == 1 and moca >= 25)
        if (cog_2a or has_subthresh or is_on_meds
                or has_rbd or has_hyposmic):
            return 'Stage 2A', fb

        return 'Stage 1A', fb  # S+ D- no signs default

    # ──────────────────────────────────────────────────────────
    # S+ D+ from here
    # ──────────────────────────────────────────────────────────

    # STAGE 1B: S+ D+, completely asymptomatic
    # COG1=0 AND no subthreshold AND not on meds
    # AND no RBD AND not hyposmic
    # ──────────────────────────────────────────────────────────
    if (pd.notna(cog1) and cog1 == 0
            and not has_subthresh and not is_on_meds
            and not has_rbd and not has_hyposmic):
        return 'Stage 1B', fb

    # ──────────────────────────────────────────────────────────
    # Define criteria for stages 3+ before assigning 2B
    # ──────────────────────────────────────────────────────────
    cog_3 = (pd.notna(cog1) and pd.notna(moca) and
             ((cog1 == 1 and moca <= 24) or
              (cog1 == 2 and moca >= 25)))
    mot_3 = (pd.notna(part2) and 3 <= part2 <= 13
             and (has_subthresh or is_on_meds))

    cog_4 = (pd.notna(cog1) and pd.notna(moca) and
             ((cog1 == 2 and moca <= 24) or
              (cog1 == 3 and moca >= 25)))
    mot_4 = (pd.notna(part2) and 14 <= part2 <= 26)
    non_4 = (pd.notna(part1_ex) and 13 <= part1_ex <= 24)

    cog_5 = (pd.notna(cog1) and pd.notna(moca) and
             ((cog1 == 3 and moca <= 24) or
              (cog1 == 4 and moca >= 25)))
    mot_5 = (pd.notna(part2) and 27 <= part2 <= 39)
    non_5 = (pd.notna(part1_ex) and 25 <= part1_ex <= 36)

    cog_6 = (pd.notna(cog1) and pd.notna(moca)
              and cog1 == 4 and moca <= 24)
    mot_6 = (pd.notna(part2) and part2 >= 40)
    non_6 = (pd.notna(part1_ex) and part1_ex >= 37)

    # ──────────────────────────────────────────────────────────
    # STAGE 2B: S+ D+, subtle signs, no functional impairment
    # COG1=1 AND MoCA>=25 OR has subthreshold OR on meds
    # OR has RBD OR hyposmic
    # Must NOT meet Stage 3+ criteria
    # ──────────────────────────────────────────────────────────
    cog_2b = (pd.notna(cog1) and pd.notna(moca)
              and cog1 == 1 and moca >= 25)
    stage_2b = (cog_2b or has_subthresh or is_on_meds
                or has_rbd or has_hyposmic)
    meets_3plus = (cog_3 or mot_3 or cog_4 or mot_4 or
                   non_4 or cog_5 or mot_5 or non_5 or
                   cog_6 or mot_6 or non_6)

    if stage_2b and not meets_3plus:
        return 'Stage 2B', fb

    # ──────────────────────────────────────────────────────────
    # STAGE 3: S+ D+
    # Cognitive: COG1=1 AND MoCA<=24 OR COG1=2 AND MoCA>=25
    # Motor:     UPDRS-II=3-13 AND (subthreshold OR on meds)
    # ──────────────────────────────────────────────────────────
    if cog_3 or mot_3:
        return 'Stage 3', fb

    # ──────────────────────────────────────────────────────────
    # STAGE 4: S+ D+
    # Cognitive: COG1=2 AND MoCA<=24 OR COG1=3 AND MoCA>=25
    # Motor:     UPDRS-II=14-26
    # Non-motor: UPDRS-I(excl 1.1)=13-24
    # ──────────────────────────────────────────────────────────
    if cog_4 or mot_4 or non_4:
        return 'Stage 4', fb

    # ──────────────────────────────────────────────────────────
    # STAGE 5: S+ D+
    # Cognitive: COG1=3 AND MoCA<=24 OR COG1=4 AND MoCA>=25
    # Motor:     UPDRS-II=27-39
    # Non-motor: UPDRS-I(excl 1.1)=25-36
    # ──────────────────────────────────────────────────────────
    if cog_5 or mot_5 or non_5:
        return 'Stage 5', fb

    # ──────────────────────────────────────────────────────────
    # STAGE 6: S+ D+
    # Cognitive: COG1=4 AND MoCA<=24
    # Motor:     UPDRS-II>=40
    # Non-motor: UPDRS-I(excl 1.1)>=37
    # ──────────────────────────────────────────────────────────
    if cog_6 or mot_6 or non_6:
        return 'Stage 6', fb

    return 'Indeterminate', fb





```

```python
# ══════════════════════════════════════════════════════════════
# PART 3 — APPLY STAGING
# ══════════════════════════════════════════════════════════════
pd_mask = (main_df_final['COHORT_DEFINITION']
           == "Parkinson's Disease")
hc_mask = ~pd_mask

print("Running NSD-ISS staging on PD patients...")

results = (main_df_final[pd_mask]
           .apply(lambda row: pd.Series(
               assign_nsd_iss_ppmi(row)), axis=1))
results.columns = ['NSD_ISS_Stage', 'fallback_notes']

main_df_final.loc[pd_mask, 'NSD_ISS_Stage']  = (
    results['NSD_ISS_Stage'].values)
main_df_final.loc[pd_mask, 'fallback_notes'] = (
    results['fallback_notes'].values)

main_df_final.loc[hc_mask, 'NSD_ISS_Stage']  = 'HC'
main_df_final.loc[hc_mask, 'fallback_notes'] = ''
```

```python
from scipy import stats

# List of genetic mutation columns that might contain non-numeric values
genetic_mutation_cols = ['LRRK2', 'GBA', 'VPS35', 'SNCA', 'PRKN', 'PARK7', 'PINK1']

# Convert genetic mutation columns to binary (0 or 1)
# 0 for no mutation, 1 for any mutation string
for col in genetic_mutation_cols:
    if col in df_new.columns:
        df_new[col] = df_new[col].astype(str).str.strip().apply(
            lambda x: 0 if x == '0' or x == 'nan' or x == '' else 1
        )
        df_new[col] = pd.to_numeric(df_new[col], errors='coerce')

results = []

# Spearman for continuous + ordinal
for col in continuous + ordinal:
    tmp = df_new[['stage_ord', col]].dropna()
    if len(tmp) < 10:
        continue
    rho, pval = stats.spearmanr(tmp['stage_ord'], tmp[col])
    results.append({'feature': col, 'method': 'spearman',
                    'stat': rho, 'pval': pval, 'type': 'continuous/ordinal'})

# Point-biserial for binary
for col in binary:
    tmp = df_new[['stage_ord', col]].dropna()
    if len(tmp) < 10 or tmp[col].nunique() < 2:
        continue
    r, pval = stats.pointbiserialr(tmp[col], tmp['stage_ord'])
    results.append({'feature': col, 'method': 'point_biserial',
                    'stat': r, 'pval': pval, 'type': 'binary'})

# Kruskal-Wallis for categorical
for col in categorical:
    tmp = df_new[['stage_ord', col]].dropna()
    if len(tmp) < 10 or tmp[col].nunique() < 2:
        continue
    groups = [tmp['stage_ord'][tmp[col] == v].values for v in tmp[col].unique()]
    stat, pval = stats.kruskal(*groups)
    results.append({'feature': col, 'method': 'kruskal',
                    'stat': stat, 'pval': pval, 'type': 'categorical'})

corr_df = pd.DataFrame(results).sort_values('stat', key=abs, ascending=False)
corr_df['abs_stat'] = corr_df['stat'].abs()
print(corr_df.to_string(index=False))
```

```python
# ================================================================
# CORRECT ORDER
# ================================================================

# STEP 1 — Stratify FIRST (while columns still have string values)
def stratify_genetics(row):
    if str(row['GBA']) in ['L483P', 'R159W', 'N409S/N409S']:
        return 'GBA_Severe'
    elif str(row['GBA']) in ['N409S', 'R502C']:
        return 'GBA_Mild'
    elif str(row['LRRK2']) == 'G2019S':
        return 'LRRK2_Carrier'
    elif str(row['PINK1']) == 'R492X':
        return 'PINK1_Carrier'
    else:
        return 'Idiopathic_PD_or_HC'

df_all['Genetic_Severity'] = df_all.apply(stratify_genetics, axis=1)

print("Distribution before OHE:")
print(df_all['Genetic_Severity'].value_counts())
# Now you'll see proper distribution

# STEP 2 — Convert raw genetic cols to binary AFTER stratification
for col in ['LRRK2', 'GBA', 'PINK1']:
    df_all[col] = df_all[col].apply(
        lambda x: 0 if str(x).strip() in ('0','nan','') else 1
    )

# STEP 3 — OHE with small sample protection
df_model = pd.get_dummies(
    df_all,
    columns=['Genetic_Severity'],
    drop_first=False,   # keep all for now, drop manually below
    dtype=int
)

# STEP 4 — Drop rare categories (< 10 samples)
# Note: < 3 is too lenient — after 80/20 split you might have
# only 1-2 samples in train for that class
gen_cols = [c for c in df_model.columns if 'Genetic_Severity_' in c]
for col in gen_cols:
    count = df_model[col].sum()
    if count < 10:
        print(f"⚠️  Dropping {col} — only {count} samples (too few)")
        df_model.drop(columns=[col], inplace=True)
    else:
        print(f"✅ Keeping {col} — {count} samples")

# STEP 5 — Drop reference category manually (avoid dummy trap)
# Keep: GBA_Severe, GBA_Mild, LRRK2_Carrier
# Drop: Idiopathic_PD_or_HC (reference), PINK1_Carrier (too rare)
ref_col = 'Genetic_Severity_Idiopathic_PD_or_HC'
if ref_col in df_model.columns:
    df_model.drop(columns=[ref_col], inplace=True)
    print(f"Dropped reference category: {ref_col}")

# Final check
gen_cols_final = [c for c in df_model.columns
                  if 'Genetic_Severity_' in c]
print(f"\nFinal genetic dummy columns: {gen_cols_final}")
print(df_model[gen_cols_final].sum())
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import (DataLoader, TensorDataset,
                               WeightedRandomSampler)
from sklearn.model_selection import GroupKFold
from sklearn.preprocessing import StandardScaler
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (classification_report,
                             f1_score, accuracy_score)
import numpy as np
import pandas as pd

# ================================================================
# MODEL — Clinical Excitation Network
# ================================================================
class ClinicalExcitationNetwork(nn.Module):
    def __init__(self, input_dim, embedding_dim=64,
                 hidden_dim=64, num_classes=4, dropout=0.2):
        super().__init__()

        # Excitation branch — per-feature attention (0→1)
        self.excitation = nn.Sequential(
            nn.Linear(input_dim, input_dim // 2),
            nn.ReLU(),
            nn.Linear(input_dim // 2, input_dim),
            nn.Sigmoid()
        )

        # Embedding branch — learn non-linear interactions
        self.embedding = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, embedding_dim),
            nn.BatchNorm1d(embedding_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
        )

        # Classifier head — 4-class output
        self.classifier = nn.Sequential(
            nn.Linear(embedding_dim, 32),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(32, num_classes)
        )

    def forward(self, x):
        weights    = self.excitation(x)      # (batch, input_dim)
        x_weighted = x * weights             # (batch, input_dim)
        embedding  = self.embedding(x_weighted)  # (batch, 64)
        logits     = self.classifier(embedding)  # (batch, 4)
        return logits, embedding, weights


# ================================================================
# TRAINING UTILITIES
# ================================================================
def train_one_epoch(model, loader, optimizer,
                    criterion, device):
    model.train()
    total_loss, correct, total = 0, 0, 0
    for X_batch, y_batch in loader:
        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)
        optimizer.zero_grad()
        logits, _, _ = model(X_batch)
        loss = criterion(logits, y_batch)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * len(y_batch)
        preds       = logits.argmax(dim=1)
        correct    += (preds == y_batch).sum().item()
        total      += len(y_batch)
    return total_loss / total, correct / total


def evaluate(model, loader, criterion, device):
    model.eval()
    total_loss, correct, total = 0, 0, 0
    all_preds, all_labels = [], []
    with torch.no_grad():
        for X_batch, y_batch in loader:
            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)
            logits, _, _ = model(X_batch)
            loss = criterion(logits, y_batch)
            total_loss += loss.item() * len(y_batch)
            preds       = logits.argmax(dim=1)
            correct    += (preds == y_batch).sum().item()
            total      += len(y_batch)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(y_batch.cpu().numpy())
    avg_loss = total_loss / total
    acc      = correct / total
    f1       = f1_score(all_labels, all_preds,
                        average='macro', zero_division=0)
    return avg_loss, acc, f1, all_preds, all_labels


# ================================================================
# HYPERPARAMETERS
# ================================================================
device = torch.device('cuda' if torch.cuda.is_available()
                      else 'cpu')
print(f"Device: {device}")

INPUT_DIM     = len(all_features)
EMBEDDING_DIM = 64
HIDDEN_DIM    = 64
NUM_CLASSES   = 4
DROPOUT       = 0.2
WEIGHT_DECAY  = 1e-3
LR            = 3e-4
EPOCHS        = 100
BATCH_SIZE    = 32
PATIENCE      = 20

labels_map = {0:'HC', 1:'Early', 2:'Moderate', 3:'Advanced'}
gkf        = GroupKFold(n_splits=5)
fold_results = []

print(f"\nInput dim:  {INPUT_DIM}")
print(f"Embedding:  {EMBEDDING_DIM}")
print(f"Hidden:     {HIDDEN_DIM}")
print(f"Dropout:    {DROPOUT}")
print(f"LR:         {LR}")
print(f"Weight dec: {WEIGHT_DECAY}")
print(f"Batch size: {BATCH_SIZE}")
print(f"Patience:   {PATIENCE}")

# ================================================================
# 5-FOLD CROSS-VALIDATION
# ================================================================
for fold, (train_idx, val_idx) in enumerate(
        gkf.split(X, y, groups=groups)):

    print(f"\n{'='*55}")
    print(f"FOLD {fold+1}/5")
    print(f"{'='*55}")

    # ── Split ─────────────────────────────────────────────────
    X_train_np = X.iloc[train_idx].values.copy()
    X_val_np   = X.iloc[val_idx].values.copy()
    y_train_np = y[train_idx]
    y_val_np   = y[val_idx]

    # ── Scale continuous only (fit on train) ──────────────────
    cont_idx = [all_features.index(c)
                for c in continuous_features]
    scaler = StandardScaler()
    X_train_np[:, cont_idx] = scaler.fit_transform(
        X_train_np[:, cont_idx])
    X_val_np[:, cont_idx]   = scaler.transform(
        X_val_np[:, cont_idx])

    # ── Class weights for weighted loss ───────────────────────
    cw = compute_class_weight(
        class_weight='balanced',
        classes=np.unique(y_train_np),
        y=y_train_np)
    cw_tensor = torch.tensor(
        cw, dtype=torch.float).to(device)
    criterion = nn.CrossEntropyLoss(weight=cw_tensor)

    print(f"  Class weights:")
    for i, (w, n) in enumerate(
            zip(cw, np.bincount(y_train_np))):
        print(f"    {labels_map[i]:<10} "
              f"n={n:3d}  weight={w:.3f}")

    # ── Weighted sampler (sqrt to avoid overcompensation) ─────
    sample_weights = torch.tensor(
        [np.sqrt(cw[label]) for label in y_train_np],
        dtype=torch.float)
    sampler = WeightedRandomSampler(
        weights     = sample_weights,
        num_samples = len(sample_weights),
        replacement = True)

    # ── Tensors ───────────────────────────────────────────────
    X_tr = torch.tensor(X_train_np, dtype=torch.float32)
    y_tr = torch.tensor(y_train_np, dtype=torch.long)
    X_vl = torch.tensor(X_val_np,   dtype=torch.float32)
    y_vl = torch.tensor(y_val_np,   dtype=torch.long)

    # ── DataLoaders ───────────────────────────────────────────
    train_loader = DataLoader(
        TensorDataset(X_tr, y_tr),
        batch_size = BATCH_SIZE,
        sampler    = sampler)       # no shuffle with sampler
    val_loader = DataLoader(
        TensorDataset(X_vl, y_vl),
        batch_size = BATCH_SIZE,
        shuffle    = False)

    # ── Model ─────────────────────────────────────────────────
    model = ClinicalExcitationNetwork(
        input_dim     = INPUT_DIM,
        embedding_dim = EMBEDDING_DIM,
        hidden_dim    = HIDDEN_DIM,
        num_classes   = NUM_CLASSES,
        dropout       = DROPOUT,
    ).to(device)

    # ── Optimizer — uses WEIGHT_DECAY variable ────────────────
    optimizer = optim.Adam(
        model.parameters(),
        lr           = LR,
        weight_decay = WEIGHT_DECAY)  # ← fixed

    # ── Scheduler — mode='max' because monitoring F1 ─────────
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode     = 'max',   # ← fixed (F1 higher = better)
        patience = 10,
        factor   = 0.5)

    # ── Training loop with early stopping ────────────────────
    best_val_f1  = 0.0
    best_epoch   = 0
    patience_cnt = 0
    best_state   = None
    history      = []

    for epoch in range(EPOCHS):
        tr_loss, tr_acc = train_one_epoch(
            model, train_loader,
            optimizer, criterion, device)
        vl_loss, vl_acc, vl_f1, _, _ = evaluate(
            model, val_loader, criterion, device)

        scheduler.step(vl_f1)   # ← step on F1, not loss

        history.append({
            'epoch':   epoch + 1,
            'tr_loss': tr_loss, 'tr_acc': tr_acc,
            'vl_loss': vl_loss, 'vl_acc': vl_acc,
            'vl_f1':   vl_f1,
        })

        # Early stopping on macro F1
        if vl_f1 > best_val_f1:
            best_val_f1  = vl_f1
            best_epoch   = epoch + 1
            patience_cnt = 0
            best_state   = {
                k: v.clone()
                for k, v in model.state_dict().items()}
        else:
            patience_cnt += 1

        if patience_cnt >= PATIENCE:
            print(f"  Early stop ep={epoch+1} "
                  f"best={best_epoch}")
            break

        if (epoch + 1) % 10 == 0:
            print(f"  Ep {epoch+1:3d} | "
                  f"tr={tr_loss:.4f}/{tr_acc:.3f} | "
                  f"vl={vl_loss:.4f}/{vl_acc:.3f} "
                  f"f1={vl_f1:.3f}")

    # ── Load best weights and evaluate ───────────────────────
    model.load_state_dict(best_state)
    _, val_acc, val_f1, preds, labels_true = evaluate(
        model, val_loader, criterion, device)

    print(f"\n  Best epoch:   {best_epoch}")
    print(f"  Val Acc:      {val_acc:.4f}")
    print(f"  Val F1 macro: {val_f1:.4f}")
    print(f"\n  Classification Report:")
    print(classification_report(
        labels_true, preds,
        target_names=list(labels_map.values()),
        zero_division=0))

    print(f"  Val class distribution:")
    for k, v in zip(*np.unique(y_val_np,
                                return_counts=True)):
        print(f"    {labels_map[k]:<10}: {v}")

    fold_results.append({
        'fold':       fold + 1,
        'val_acc':    val_acc,
        'val_f1':     val_f1,
        'best_epoch': best_epoch,
        'preds':      preds,
        'labels':     labels_true,
        'history':    history,
        'scaler':     scaler,
    })

# ================================================================
# FINAL SUMMARY
# ================================================================
print(f"\n{'='*55}")
print("5-FOLD CROSS-VALIDATION SUMMARY")
print(f"{'='*55}")
accs = [r['val_acc'] for r in fold_results]
f1s  = [r['val_f1']  for r in fold_results]

for r in fold_results:
    print(f"  Fold {r['fold']}: "
          f"Acc={r['val_acc']:.4f}  "
          f"F1={r['val_f1']:.4f}  "
          f"Best_ep={r['best_epoch']}")

print(f"\n  Mean Acc: {np.mean(accs):.4f} "
      f"± {np.std(accs):.4f}")
print(f"  Mean F1:  {np.mean(f1s):.4f} "
      f"± {np.std(f1s):.4f}")

# ================================================================
# SAVE BEST FOLD MODEL
# ================================================================
best_fold_idx = int(np.argmax(f1s))
best_fold     = fold_results[best_fold_idx]
best_scaler   = best_fold['scaler']

print(f"\nBest fold: {best_fold['fold']} "
      f"(F1={best_fold['val_f1']:.4f})")

torch.save({
    'model_state':         best_state,
    'input_dim':           INPUT_DIM,
    'embedding_dim':       EMBEDDING_DIM,
    'hidden_dim':          HIDDEN_DIM,
    'num_classes':         NUM_CLASSES,
    'dropout':             DROPOUT,
    'continuous_features': continuous_features,
    'binary_features':     binary_features,
    'all_features':        all_features,
    'labels_map':          labels_map,
    'class_weights':       cw.tolist(),
    'cont_idx':            cont_idx,
}, 'clinical_excitation_model.pt')

import joblib
joblib.dump(best_scaler, 'clinical_scaler.pkl')

print("✅ Saved: clinical_excitation_model.pt")
print("✅ Saved: clinical_scaler.pkl")

# ================================================================
# LEAKAGE CHECK (run after training as sanity check)
# ================================================================
print(f"\n{'='*55}")
print("HC vs PD leakage check")
print(f"{'='*55}")
check_cols = [c for c in [
    'total_LEDD', 'motor_composite', 'DAT_measured',
    'high_ledd', 'ledd_per_motor', 'axial_score',
    'total_rigidity', 'NP3BRADY', 'NP3RIGN',
    'Part1', 'Part1P', 'RBD_score'
] if c in X.columns]

print(f"\n{'Feature':<22} {'HC mean':>10} "
      f"{'PD mean':>10} {'Flag':>10}")
print("-"*55)
for col in check_cols:
    hc_mean = X[y==0][col].mean()
    pd_mean = X[y!=0][col].mean()
    ratio   = abs(pd_mean - hc_mean) / (abs(hc_mean) + 1e-6)
    flag    = "🔴 LEAKY" if ratio > 5 else (
              "⚠️  high"  if ratio > 2 else "✅ ok")
    print(f"  {col:<20} {hc_mean:>10.3f} "
          f"{pd_mean:>10.3f} {flag:>10}")
```

```python
import os
import joblib
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, f1_score
from sklearn.utils.class_weight import compute_class_weight
import numpy as np
import pandas as pd

# # ================================================================
# # FEATURE LISTS
# # ================================================================
# continuous_features = [
#     'motor_composite', 'total_rigidity', 'axial_score',
#     'rigidity_asymmetry',
#     'DAT_STRIATAL_SBR', 'nfl_log', 'ptau_ttau_ratio',
#     'pTau', 'tTau',
#     'RBD_score', 'total_LEDD', 'ledd_per_motor',
#     'MRI_Age', 'Part1', 'Part1P',
#     'SDMTOTAL', 'MCAVFNUM', 'MCASER7', 'MCASNTNC', 'MCAABSTR'
# ]

# binary_features = [
#     'high_ledd', 'DAT_measured',
#     'Genetic_Severity_LRRK2_Carrier', 'GBA', 'PINK1',
#     'MCAALTTM', 'MCACUBE', 'MCACLCKC', 'MCACLCKN', 'MCACLCKH',
#     'MCARHINO', 'MCACAMEL', 'MCAFDS', 'MCABDS', 'MCAVIGIL'
# ]

# all_features = continuous_features + binary_features

# ================================================================
# MODEL
# ================================================================
class ClinicalExcitationNetwork(nn.Module):
    def __init__(self, input_dim, embedding_dim=64,
                 hidden_dim=64, num_classes=4, dropout=0.2):
        super().__init__()

        self.excitation = nn.Sequential(
            nn.Linear(input_dim, input_dim // 2),
            nn.ReLU(),
            nn.Linear(input_dim // 2, input_dim),
            nn.Sigmoid()
        )
        self.embedding = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, embedding_dim),
            nn.BatchNorm1d(embedding_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
        )
        self.classifier = nn.Sequential(
            nn.Linear(embedding_dim, 32),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(32, num_classes)
        )

    def forward(self, x):
        weights    = self.excitation(x)
        x_weighted = x * weights
        embedding  = self.embedding(x_weighted)
        logits     = self.classifier(embedding)
        return logits, embedding, weights


# ================================================================
# TRAINING UTILITIES
# ================================================================
def train_one_epoch(model, loader, optimizer,
                    criterion, device):
    model.train()
    total_loss, correct, total = 0, 0, 0
    for X_batch, y_batch in loader:
        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)
        optimizer.zero_grad()
        logits, _, _ = model(X_batch)
        loss = criterion(logits, y_batch)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * len(y_batch)
        preds       = logits.argmax(dim=1)
        correct    += (preds == y_batch).sum().item()
        total      += len(y_batch)
    return total_loss / total, correct / total


def evaluate(model, loader, criterion, device):
    model.eval()
    total_loss, correct, total = 0, 0, 0
    all_preds, all_labels = [], []
    with torch.no_grad():
        for X_batch, y_batch in loader:
            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)
            logits, _, _ = model(X_batch)
            loss = criterion(logits, y_batch)
            total_loss += loss.item() * len(y_batch)
            preds       = logits.argmax(dim=1)
            correct    += (preds == y_batch).sum().item()
            total      += len(y_batch)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(y_batch.cpu().numpy())
    avg_loss = total_loss / total
    acc      = correct / total
    f1       = f1_score(all_labels, all_preds,
                        average='macro', zero_division=0)
    return avg_loss, acc, f1, all_preds, all_labels


# ================================================================
# HYPERPARAMETERS
# ================================================================
device = torch.device('cuda' if torch.cuda.is_available()
                      else 'cpu')
print(f"Device: {device}")

# Build X and y
X      = df_eng[all_features].copy().astype(float)
y      = df_eng['stage_merged'].values
groups = df_eng['PATNO'].values

INPUT_DIM     = len(all_features)
EMBEDDING_DIM = 64
HIDDEN_DIM    = 64
NUM_CLASSES   = 4
DROPOUT       = 0.2
WEIGHT_DECAY  = 1e-3
LR            = 0.0003
EPOCHS        = 100
BATCH_SIZE    = 32
PATIENCE      = 20

labels_map = {0:'HC', 1:'Early', 2:'Moderate', 3:'Advanced'}

print(f"Input dim:    {INPUT_DIM}")
print(f"Embedding:    {EMBEDDING_DIM}")
print(f"Hidden:       {HIDDEN_DIM}")
print(f"Dropout:      {DROPOUT}")
print(f"LR:           {LR}")
print(f"Weight decay: {WEIGHT_DECAY}")
print(f"Batch size:   {BATCH_SIZE}")
print(f"Patience:     {PATIENCE}")

labels = {0:'HC', 1:'Early', 2:'Moderate', 3:'Advanced'}
print(f"\ny distribution:")
for k, v in zip(*np.unique(y, return_counts=True)):
    print(f"  {k} ({labels[k]}): {v}")

# ================================================================
# INITIALIZE OOF EMBEDDINGS + RESULTS STORE
# ================================================================
# Out-of-fold embeddings — each patient's embedding comes
# from a model that never saw them during training
oof_embeddings = np.zeros((len(X), EMBEDDING_DIM))

sgkf         = StratifiedGroupKFold(
    n_splits=5, shuffle=True, random_state=42)
fold_results = []
cont_idx     = [all_features.index(c)
                for c in continuous_features]

# ================================================================
# 5-FOLD CROSS-VALIDATION LOOP
# ================================================================
for fold, (train_idx, val_idx) in enumerate(
        sgkf.split(X, y, groups=groups)):

    print(f"\n{'='*55}")
    print(f"FOLD {fold+1}/5")
    print(f"{'='*55}")

    # ── Split ─────────────────────────────────────────────────
    X_train_np = X.iloc[train_idx].values.copy()
    X_val_np   = X.iloc[val_idx].values.copy()
    y_train_np = y[train_idx]
    y_val_np   = y[val_idx]

    # ── Scale continuous only (fit on train) ──────────────────
    scaler = StandardScaler()
    X_train_np[:, cont_idx] = scaler.fit_transform(
        X_train_np[:, cont_idx])
    X_val_np[:, cont_idx]   = scaler.transform(
        X_val_np[:, cont_idx])

    # ── Smoothed sqrt class weights (single weighting) ────────
    # 1/sqrt(count) then normalize so minimum weight = 1.0
    # Gentler than full inverse — prevents Moderate collapse
    class_counts = np.bincount(y_train_np)
    smoothed_cw  = 1.0 / np.sqrt(class_counts)
    smoothed_cw  = smoothed_cw / np.min(smoothed_cw)

    cw_tensor = torch.tensor(
        smoothed_cw, dtype=torch.float32).to(device)
    criterion = nn.CrossEntropyLoss(weight=cw_tensor)

    print(f"  Class weights (sqrt smoothed):")
    for i, (w, n) in enumerate(
            zip(smoothed_cw, class_counts)):
        print(f"    {labels_map[i]:<10} "
              f"n={n:3d}  weight={w:.3f}")

    # ── Tensors ───────────────────────────────────────────────
    X_tr = torch.tensor(X_train_np, dtype=torch.float32)
    y_tr = torch.tensor(y_train_np, dtype=torch.long)
    X_vl = torch.tensor(X_val_np,   dtype=torch.float32)
    y_vl = torch.tensor(y_val_np,   dtype=torch.long)

    # ── DataLoaders — shuffle only, no sampler ────────────────
    train_loader = DataLoader(
        TensorDataset(X_tr, y_tr),
        batch_size = BATCH_SIZE,
        shuffle    = True)
    val_loader = DataLoader(
        TensorDataset(X_vl, y_vl),
        batch_size = BATCH_SIZE,
        shuffle    = False)

    # ── Model ─────────────────────────────────────────────────
    model = ClinicalExcitationNetwork(
        input_dim     = INPUT_DIM,
        embedding_dim = EMBEDDING_DIM,
        hidden_dim    = HIDDEN_DIM,
        num_classes   = NUM_CLASSES,
        dropout       = DROPOUT,
    ).to(device)

    optimizer = optim.Adam(
        model.parameters(),
        lr           = LR,
        weight_decay = WEIGHT_DECAY)

    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode     = 'max',   # monitor F1 — higher is better
        patience = 10,
        factor   = 0.5)

    # ── Training loop with early stopping ────────────────────
    best_val_f1  = 0.0
    best_epoch   = 0
    patience_cnt = 0
    best_state   = None
    history      = []

    for epoch in range(EPOCHS):
        tr_loss, tr_acc = train_one_epoch(
            model, train_loader,
            optimizer, criterion, device)
        vl_loss, vl_acc, vl_f1, _, _ = evaluate(
            model, val_loader, criterion, device)

        scheduler.step(vl_f1)

        history.append({
            'epoch':   epoch + 1,
            'tr_loss': tr_loss, 'tr_acc': tr_acc,
            'vl_loss': vl_loss, 'vl_acc': vl_acc,
            'vl_f1':   vl_f1,
        })

        if vl_f1 > best_val_f1:
            best_val_f1  = vl_f1
            best_epoch   = epoch + 1
            patience_cnt = 0
            best_state   = {
                k: v.clone()
                for k, v in model.state_dict().items()}
        else:
            patience_cnt += 1

        if patience_cnt >= PATIENCE:
            print(f"  Early stop ep={epoch+1} "
                  f"best={best_epoch}")
            break

        if (epoch + 1) % 10 == 0:
            print(f"  Ep {epoch+1:3d} | "
                  f"tr={tr_loss:.4f}/{tr_acc:.3f} | "
                  f"vl={vl_loss:.4f}/{vl_acc:.3f} "
                  f"f1={vl_f1:.3f}")

    # ── Load best weights ─────────────────────────────────────
    model.load_state_dict(best_state)
    _, val_acc, val_f1, preds, labels_true = evaluate(
        model, val_loader, criterion, device)

    # ── Extract OOF embeddings (no leakage) ───────────────────
    # Val patients were never seen during training
    model.eval()
    with torch.no_grad():
        _, fold_embs, _ = model(X_vl.to(device))
        oof_embeddings[val_idx] = fold_embs.cpu().numpy()

    print(f"\n  Best epoch:   {best_epoch}")
    print(f"  Val Acc:      {val_acc:.4f}")
    print(f"  Val F1 macro: {val_f1:.4f}")
    print(f"\n  Classification Report:")
    print(classification_report(
        labels_true, preds,
        target_names=list(labels_map.values()),
        zero_division=0))

    print(f"  Val class distribution:")
    for k, v in zip(*np.unique(y_val_np,
                                return_counts=True)):
        print(f"    {labels_map[k]:<10}: {v}")

    # ── Store fold results including best_state ───────────────
    fold_results.append({
        'fold':        fold + 1,
        'val_acc':     val_acc,
        'val_f1':      val_f1,
        'best_epoch':  best_epoch,
        'preds':       preds,
        'labels':      labels_true,
        'history':     history,
        'scaler':      scaler,
        'best_state':  best_state,   # ← per-fold best weights
        'smoothed_cw': smoothed_cw,  # ← per-fold weights
    })

# ================================================================
# FINAL SUMMARY
# ================================================================
print(f"\n{'='*55}")
print("5-FOLD CROSS-VALIDATION SUMMARY")
print(f"{'='*55}")
accs = [r['val_acc']   for r in fold_results]
f1s  = [r['val_f1']    for r in fold_results]
eps  = [r['best_epoch'] for r in fold_results]

for r in fold_results:
    print(f"  Fold {r['fold']}: "
          f"Acc={r['val_acc']:.4f}  "
          f"F1={r['val_f1']:.4f}  "
          f"Best_ep={r['best_epoch']}")

print(f"\n  Mean Acc: {np.mean(accs):.4f} "
      f"± {np.std(accs):.4f}")
print(f"  Mean F1:  {np.mean(f1s):.4f} "
      f"± {np.std(f1s):.4f}")

avg_best_epoch = int(np.mean(eps))
print(f"  Avg best epoch: {avg_best_epoch}")

# ================================================================
# SAVE BEST FOLD MODEL
# ================================================================
# Use best fold's state — not last fold's state
best_fold_idx = int(np.argmax(f1s))
best_fold     = fold_results[best_fold_idx]
best_state    = best_fold['best_state']   # ← correct fold
best_scaler   = best_fold['scaler']

print(f"\nBest fold: {best_fold['fold']} "
      f"(F1={best_fold['val_f1']:.4f})")

# Class weights computed on full dataset for inference
final_counts = np.bincount(y)
final_cw     = 1.0 / np.sqrt(final_counts)
final_cw     = final_cw / np.min(final_cw)

print(f"\nFull dataset class weights:")
for i, (w, n) in enumerate(zip(final_cw, final_counts)):
    print(f"  {labels_map[i]:<10} n={n:3d}  weight={w:.3f}")

torch.save({
    'model_state':         best_state,
    'input_dim':           INPUT_DIM,
    'embedding_dim':       EMBEDDING_DIM,
    'hidden_dim':          HIDDEN_DIM,
    'num_classes':         NUM_CLASSES,
    'dropout':             DROPOUT,
    'continuous_features': continuous_features,
    'binary_features':     binary_features,
    'all_features':        all_features,
    'labels_map':          labels_map,
    'class_weights':       final_cw.tolist(),
    'cont_idx':            cont_idx,
    'cv_mean_f1':          float(np.mean(f1s)),
    'cv_std_f1':           float(np.std(f1s)),
    'best_fold':           best_fold['fold'],
    'avg_best_epoch':      avg_best_epoch,
}, 'clinical_excitation_model.pt')

joblib.dump(best_scaler, 'clinical_scaler.pkl')

# ================================================================
# SAVE OOF EMBEDDINGS WITH PATIENT IDs
# ================================================================
# Verify all rows filled — no zeros remaining
zeros_remaining = (oof_embeddings.sum(axis=1) == 0).sum()
print(f"\nOOF zeros remaining: {zeros_remaining} "
      f"← must be 0")

# Attach patient identifiers
ref_cols = [
    'PATNO', 'EVENT_ID',
    'Image_ID_T1', 'Image_ID_T2',
    'NSD_ISS_Stage', 'label_binary', 'stage_merged'
]
ref_cols = [c for c in ref_cols if c in df_eng.columns]
ref_df   = df_eng[ref_cols].reset_index(drop=True)

emb_cols = [f'emb_{i}' for i in range(EMBEDDING_DIM)]
emb_df   = pd.DataFrame(oof_embeddings, columns=emb_cols)
final_df = pd.concat([ref_df, emb_df], axis=1)

# Save both formats
final_df.to_csv('clinical_embeddings_oof.csv', index=False)
np.save('clinical_embeddings_oof.npy', oof_embeddings)

print(f"\n✅ Saved: clinical_excitation_model.pt")
print(f"✅ Saved: clinical_scaler.pkl")
print(f"✅ Saved: clinical_embeddings_oof.csv "
      f"({final_df.shape})")
print(f"✅ Saved: clinical_embeddings_oof.npy "
      f"({oof_embeddings.shape})")

# ================================================================
# VERIFICATION
# ================================================================
print(f"\n{'='*55}")
print("VERIFICATION")
print(f"{'='*55}")
print(f"  Unique patients:     {final_df['PATNO'].nunique()}")
print(f"  Total visits:        {len(final_df)}")
print(f"  T1 images linked:    "
      f"{final_df['Image_ID_T1'].notna().sum()}")
print(f"  T2 images linked:    "
      f"{final_df['Image_ID_T2'].notna().sum()}")
print(f"  Embedding NaN:       "
      f"{np.isnan(oof_embeddings).any()}")
print(f"  Embedding shape:     {oof_embeddings.shape}")
print(f"  CV Mean F1:          {np.mean(f1s):.4f} "
      f"± {np.std(f1s):.4f}")

print(f"\n  Stage distribution in embeddings:")
for k, v in zip(*np.unique(
        final_df['stage_merged'], return_counts=True)):
    print(f"    {labels_map[k]:<10}: {v}")

# ================================================================
# LEAKAGE CHECK
# ================================================================
print(f"\n{'='*55}")
print("HC vs PD leakage check")
print(f"{'='*55}")
check_cols = [c for c in [
     'motor_composite', 'DAT_measured',

    'total_rigidity', 'Part1', 'Part1P', 'RBD_score'
] if c in X.columns]

print(f"\n{'Feature':<22} {'HC mean':>10} "
      f"{'PD mean':>10} {'Flag':>10}")
print("-"*57)
for col in check_cols:
    hc_mean = X[y==0][col].mean()
    pd_mean = X[y!=0][col].mean()
    ratio   = abs(pd_mean - hc_mean) / (abs(hc_mean) + 1e-6)
    flag    = "🔴 LEAKY" if ratio > 5 else (
              "⚠️  high"  if ratio > 2 else "✅ ok")
    print(f"  {col:<20} {hc_mean:>10.3f} "
          f"{pd_mean:>10.3f} {flag:>10}")
```

```python
import os
import joblib
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, f1_score
from sklearn.utils.class_weight import compute_class_weight
import numpy as np
import pandas as pd

# # ================================================================
# # FEATURE LISTS
# # ================================================================
# continuous_features = [
#     'motor_composite', 'total_rigidity', 'axial_score',
#     'rigidity_asymmetry',
#     'DAT_STRIATAL_SBR', 'nfl_log', 'ptau_ttau_ratio',
#     'pTau', 'tTau',
#     'RBD_score', 'total_LEDD', 'ledd_per_motor',
#     'MRI_Age', 'Part1', 'Part1P',
#     'SDMTOTAL', 'MCAVFNUM', 'MCASER7', 'MCASNTNC', 'MCAABSTR'
# ]

# binary_features = [
#     'high_ledd', 'DAT_measured',
#     'Genetic_Severity_LRRK2_Carrier', 'GBA', 'PINK1',
#     'MCAALTTM', 'MCACUBE', 'MCACLCKC', 'MCACLCKN', 'MCACLCKH',
#     'MCARHINO', 'MCACAMEL', 'MCAFDS', 'MCABDS', 'MCAVIGIL'
# ]

# all_features = continuous_features + binary_features

# ================================================================
# MODEL
# ================================================================
class ClinicalExcitationNetwork(nn.Module):
    def __init__(self, input_dim, embedding_dim=64,
                 hidden_dim=64, num_classes=3, dropout=0.2):
        super().__init__()

        self.excitation = nn.Sequential(
            nn.Linear(input_dim, input_dim // 2),
            nn.ReLU(),
            nn.Linear(input_dim // 2, input_dim),
            nn.Sigmoid()
        )
        self.embedding = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, embedding_dim),
            nn.BatchNorm1d(embedding_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
        )
        self.classifier = nn.Sequential(
            nn.Linear(embedding_dim, 32),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(32, num_classes)
        )

    def forward(self, x):
        weights    = self.excitation(x)
        x_weighted = x * weights
        embedding  = self.embedding(x_weighted)
        logits     = self.classifier(embedding)
        return logits, embedding, weights


# ================================================================
# TRAINING UTILITIES
# ================================================================
def train_one_epoch(model, loader, optimizer,
                    criterion, device):
    model.train()
    total_loss, correct, total = 0, 0, 0
    for X_batch, y_batch in loader:
        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)
        optimizer.zero_grad()
        logits, _, _ = model(X_batch)
        loss = criterion(logits, y_batch)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * len(y_batch)
        preds       = logits.argmax(dim=1)
        correct    += (preds == y_batch).sum().item()
        total      += len(y_batch)
    return total_loss / total, correct / total


def evaluate(model, loader, criterion, device):
    model.eval()
    total_loss, correct, total = 0, 0, 0
    all_preds, all_labels = [], []
    with torch.no_grad():
        for X_batch, y_batch in loader:
            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)
            logits, _, _ = model(X_batch)
            loss = criterion(logits, y_batch)
            total_loss += loss.item() * len(y_batch)
            preds       = logits.argmax(dim=1)
            correct    += (preds == y_batch).sum().item()
            total      += len(y_batch)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(y_batch.cpu().numpy())
    avg_loss = total_loss / total
    acc      = correct / total
    f1       = f1_score(all_labels, all_preds,
                        average='macro', zero_division=0)
    return avg_loss, acc, f1, all_preds, all_labels


# ================================================================
# HYPERPARAMETERS
# ================================================================
device = torch.device('cuda' if torch.cuda.is_available()
                      else 'cpu')
print(f"Device: {device}")

# Build X and y
X      = df_eng[all_features].copy().astype(float)
y      = df_eng['stage_merged'].values
groups = df_eng['PATNO'].values

INPUT_DIM     = len(all_features)
EMBEDDING_DIM = 64
HIDDEN_DIM    = 64
NUM_CLASSES   = 3
DROPOUT       = 0.3
# WEIGHT_DECAY  = 1e-3
WEIGHT_DECAY  = 0.005
# LR            = 0.0003
LR            =0.001
EPOCHS        = 100
BATCH_SIZE    = 32
PATIENCE      = 20

labels_map = {0:'HC', 1:'Early', 2:'Moderate'}

print(f"Input dim:    {INPUT_DIM}")
print(f"Embedding:    {EMBEDDING_DIM}")
print(f"Hidden:       {HIDDEN_DIM}")
print(f"Dropout:      {DROPOUT}")
print(f"LR:           {LR}")
print(f"Weight decay: {WEIGHT_DECAY}")
print(f"Batch size:   {BATCH_SIZE}")
print(f"Patience:     {PATIENCE}")

labels = {0:'HC', 1:'Early', 2:'Moderate'}
print(f"\ny distribution:")
for k, v in zip(*np.unique(y, return_counts=True)):
    print(f"  {k} ({labels[k]}): {v}")

# ================================================================
# INITIALIZE OOF EMBEDDINGS + RESULTS STORE
# ================================================================
# Out-of-fold embeddings — each patient's embedding comes
# from a model that never saw them during training
oof_embeddings = np.zeros((len(X), EMBEDDING_DIM))

sgkf         = StratifiedGroupKFold(
    n_splits=5, shuffle=True, random_state=42)
fold_results = []
cont_idx     = [all_features.index(c)
                for c in continuous_features]

# ================================================================
# 5-FOLD CROSS-VALIDATION LOOP
# ================================================================
for fold, (train_idx, val_idx) in enumerate(
        sgkf.split(X, y, groups=groups)):

    print(f"\n{'='*55}")
    print(f"FOLD {fold+1}/5")
    print(f"{'='*55}")

    # ── Split ─────────────────────────────────────────────────
    X_train_np = X.iloc[train_idx].values.copy()
    X_val_np   = X.iloc[val_idx].values.copy()
    y_train_np = y[train_idx]
    y_val_np   = y[val_idx]

    # ── Scale continuous only (fit on train) ──────────────────
    scaler = StandardScaler()
    X_train_np[:, cont_idx] = scaler.fit_transform(
        X_train_np[:, cont_idx])
    X_val_np[:, cont_idx]   = scaler.transform(
        X_val_np[:, cont_idx])

    # ── Smoothed sqrt class weights (single weighting) ────────
    # 1/sqrt(count) then normalize so minimum weight = 1.0
    # Gentler than full inverse — prevents Moderate collapse
    class_counts = np.bincount(y_train_np)
    smoothed_cw  = 1.0 / np.sqrt(class_counts)
    smoothed_cw  = smoothed_cw / np.min(smoothed_cw)

    cw_tensor = torch.tensor(
        smoothed_cw, dtype=torch.float32).to(device)
    criterion = nn.CrossEntropyLoss(weight=cw_tensor)

    print(f"  Class weights (sqrt smoothed):")
    for i, (w, n) in enumerate(
            zip(smoothed_cw, class_counts)):
        print(f"    {labels_map[i]:<10} "
              f"n={n:3d}  weight={w:.3f}")

    # ── Tensors ───────────────────────────────────────────────
    X_tr = torch.tensor(X_train_np, dtype=torch.float32)
    y_tr = torch.tensor(y_train_np, dtype=torch.long)
    X_vl = torch.tensor(X_val_np,   dtype=torch.float32)
    y_vl = torch.tensor(y_val_np,   dtype=torch.long)

    # ── DataLoaders — shuffle only, no sampler ────────────────
    train_loader = DataLoader(
        TensorDataset(X_tr, y_tr),
        batch_size = BATCH_SIZE,
        shuffle    = True)
    val_loader = DataLoader(
        TensorDataset(X_vl, y_vl),
        batch_size = BATCH_SIZE,
        shuffle    = False)

    # ── Model ─────────────────────────────────────────────────
    model = ClinicalExcitationNetwork(
        input_dim     = INPUT_DIM,
        embedding_dim = EMBEDDING_DIM,
        hidden_dim    = HIDDEN_DIM,
        num_classes   = NUM_CLASSES,
        dropout       = DROPOUT,
    ).to(device)

    optimizer = optim.Adam(
        model.parameters(),
        lr           = LR,
        weight_decay = WEIGHT_DECAY)

    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode     = 'max',   # monitor F1 — higher is better
        patience = 10,
        factor   = 0.5)

    # ── Training loop with early stopping ────────────────────
    best_val_f1  = 0.0
    best_epoch   = 0
    patience_cnt = 0
    best_state   = None
    history      = []

    for epoch in range(EPOCHS):
        tr_loss, tr_acc = train_one_epoch(
            model, train_loader,
            optimizer, criterion, device)
        vl_loss, vl_acc, vl_f1, _, _ = evaluate(
            model, val_loader, criterion, device)

        scheduler.step(vl_f1)

        history.append({
            'epoch':   epoch + 1,
            'tr_loss': tr_loss, 'tr_acc': tr_acc,
            'vl_loss': vl_loss, 'vl_acc': vl_acc,
            'vl_f1':   vl_f1,
        })

        if vl_f1 > best_val_f1:
            best_val_f1  = vl_f1
            best_epoch   = epoch + 1
            patience_cnt = 0
            best_state   = {
                k: v.clone()
                for k, v in model.state_dict().items()}
        else:
            patience_cnt += 1

        if patience_cnt >= PATIENCE:
            print(f"  Early stop ep={epoch+1} "
                  f"best={best_epoch}")
            break

        if (epoch + 1) % 10 == 0:
            print(f"  Ep {epoch+1:3d} | "
                  f"tr={tr_loss:.4f}/{tr_acc:.3f} | "
                  f"vl={vl_loss:.4f}/{vl_acc:.3f} "
                  f"f1={vl_f1:.3f}")

    # ── Load best weights ─────────────────────────────────────
    model.load_state_dict(best_state)
    _, val_acc, val_f1, preds, labels_true = evaluate(
        model, val_loader, criterion, device)

    # ── Extract OOF embeddings (no leakage) ───────────────────
    # Val patients were never seen during training
    model.eval()
    with torch.no_grad():
        _, fold_embs, _ = model(X_vl.to(device))
        oof_embeddings[val_idx] = fold_embs.cpu().numpy()

    print(f"\n  Best epoch:   {best_epoch}")
    print(f"  Val Acc:      {val_acc:.4f}")
    print(f"  Val F1 macro: {val_f1:.4f}")
    print(f"\n  Classification Report:")
    print(classification_report(
        labels_true, preds,
        target_names=list(labels_map.values()),
        zero_division=0))

    print(f"  Val class distribution:")
    for k, v in zip(*np.unique(y_val_np,
                                return_counts=True)):
        print(f"    {labels_map[k]:<10}: {v}")

    # ── Store fold results including best_state ───────────────
    fold_results.append({
        'fold':        fold + 1,
        'val_acc':     val_acc,
        'val_f1':      val_f1,
        'best_epoch':  best_epoch,
        'preds':       preds,
        'labels':      labels_true,
        'history':     history,
        'scaler':      scaler,
        'best_state':  best_state,   # ← per-fold best weights
        'smoothed_cw': smoothed_cw,  # ← per-fold weights
    })

# ================================================================
# FINAL SUMMARY
# ================================================================
print(f"\n{'='*55}")
print("5-FOLD CROSS-VALIDATION SUMMARY")
print(f"{'='*55}")
accs = [r['val_acc']   for r in fold_results]
f1s  = [r['val_f1']    for r in fold_results]
eps  = [r['best_epoch'] for r in fold_results]

for r in fold_results:
    print(f"  Fold {r['fold']}: "
          f"Acc={r['val_acc']:.4f}  "
          f"F1={r['val_f1']:.4f}  "
          f"Best_ep={r['best_epoch']}")

print(f"\n  Mean Acc: {np.mean(accs):.4f} "
      f"± {np.std(accs):.4f}")
print(f"  Mean F1:  {np.mean(f1s):.4f} "
      f"± {np.std(f1s):.4f}")

avg_best_epoch = int(np.mean(eps))
print(f"  Avg best epoch: {avg_best_epoch}")

# ================================================================
# SAVE BEST FOLD MODEL
# ================================================================
# Use best fold's state — not last fold's state
best_fold_idx = int(np.argmax(f1s))
best_fold     = fold_results[best_fold_idx]
best_state    = best_fold['best_state']   # ← correct fold
best_scaler   = best_fold['scaler']

print(f"\nBest fold: {best_fold['fold']} "
      f"(F1={best_fold['val_f1']:.4f})")

# Class weights computed on full dataset for inference
final_counts = np.bincount(y, minlength=NUM_CLASSES)
final_cw = 1.0 / np.sqrt(final_counts + 1e-6)
final_cw = final_cw / np.min(final_cw)

print(f"\nFull dataset class weights:")
for i, (w, n) in enumerate(zip(final_cw, final_counts)):
    print(f"  {labels_map[i]:<10} n={n:3d}  weight={w:.3f}")

torch.save({
    'model_state':         best_state,
    'input_dim':           INPUT_DIM,
    'embedding_dim':       EMBEDDING_DIM,
    'hidden_dim':          HIDDEN_DIM,
    'num_classes':         NUM_CLASSES,
    'dropout':             DROPOUT,
    'continuous_features': continuous_features,
    'binary_features':     binary_features,
    'all_features':        all_features,
    'labels_map':          labels_map,
    'class_weights':       final_cw.tolist(),
    'cont_idx':            cont_idx,
    'cv_mean_f1':          float(np.mean(f1s)),
    'cv_std_f1':           float(np.std(f1s)),
    'best_fold':           best_fold['fold'],
    'avg_best_epoch':      avg_best_epoch,
}, 'clinical_excitation_model.pt')

joblib.dump(best_scaler, 'clinical_scaler2.pkl')

# ================================================================
# SAVE OOF EMBEDDINGS WITH PATIENT IDs
# ================================================================
# Verify all rows filled — no zeros remaining
zeros_remaining = (oof_embeddings.sum(axis=1) == 0).sum()
print(f"\nOOF zeros remaining: {zeros_remaining} "
      f"← must be 0")

# Attach patient identifiers
ref_cols = [
    'PATNO', 'EVENT_ID',
    'Image_ID_T1', 'Image_ID_T2',
    'NSD_ISS_Stage', 'label_binary', 'stage_merged'
]
ref_cols = [c for c in ref_cols if c in df_eng.columns]
ref_df   = df_eng[ref_cols].reset_index(drop=True)

emb_cols = [f'emb_{i}' for i in range(EMBEDDING_DIM)]
emb_df   = pd.DataFrame(oof_embeddings, columns=emb_cols)
final_df = pd.concat([ref_df, emb_df], axis=1)

# Save both formats
final_df.to_csv('clinical_embeddings_oof2.csv', index=False)
np.save('clinical_embeddings_oof2.npy', oof_embeddings)

print(f"\n✅ Saved: clinical_excitation_model.pt")
print(f"✅ Saved: clinical_scaler.pkl")
print(f"✅ Saved: clinical_embeddings_oof.csv "
      f"({final_df.shape})")
print(f"✅ Saved: clinical_embeddings_oof.npy "
      f"({oof_embeddings.shape})")

# ================================================================
# VERIFICATION
# ================================================================
print(f"\n{'='*55}")
print("VERIFICATION")
print(f"{'='*55}")
print(f"  Unique patients:     {final_df['PATNO'].nunique()}")
print(f"  Total visits:        {len(final_df)}")
print(f"  T1 images linked:    "
      f"{final_df['Image_ID_T1'].notna().sum()}")
print(f"  T2 images linked:    "
      f"{final_df['Image_ID_T2'].notna().sum()}")
print(f"  Embedding NaN:       "
      f"{np.isnan(oof_embeddings).any()}")
print(f"  Embedding shape:     {oof_embeddings.shape}")
print(f"  CV Mean F1:          {np.mean(f1s):.4f} "
      f"± {np.std(f1s):.4f}")

print(f"\n  Stage distribution in embeddings:")
for k, v in zip(*np.unique(
        final_df['stage_merged'], return_counts=True)):
    print(f"    {labels_map[k]:<10}: {v}")

# ================================================================
# LEAKAGE CHECK
# ================================================================
print(f"\n{'='*55}")
print("HC vs PD leakage check")
print(f"{'='*55}")
check_cols = [c for c in [
     'Part1', 'Part1P', 'RBD_score'
] if c in X.columns]

print(f"\n{'Feature':<22} {'HC mean':>10} "
      f"{'PD mean':>10} {'Flag':>10}")
print("-"*57)
for col in check_cols:
    hc_mean = X[y==0][col].mean()
    pd_mean = X[y!=0][col].mean()
    ratio   = abs(pd_mean - hc_mean) / (abs(hc_mean) + 1e-6)
    flag    = "🔴 LEAKY" if ratio > 5 else (
              "⚠️  high"  if ratio > 2 else "✅ ok")
    print(f"  {col:<20} {hc_mean:>10.3f} "
          f"{pd_mean:>10.3f} {flag:>10}")
```

```python
# ================================================================
# IMPORTS
# ================================================================
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import f1_score, classification_report
import joblib


# ================================================================
# FOCAL LOSS
# ================================================================
class FocalLoss(nn.Module):
    """
    Focal Loss = -α_t * (1 - p_t)^γ * log(p_t)

    γ (gamma) > 0  → down-weights easy confident examples,
                      forces training to focus on hard ones (Early)
    weight (α)     → per-class frequency weights (sqrt-smoothed)
    label_smoothing→ prevents overconfidence on Moderate/HC

    γ=0 is mathematically equivalent to weighted CrossEntropyLoss.
    Recommended starting point: γ=2.0 (standard Lin et al. 2017).
    Increase to 3.0 only if Early recall stays critically low.
    """
    def __init__(self, weight=None, gamma=2.0,
                 label_smoothing=0.05, reduction='mean'):
        super().__init__()
        self.weight          = weight
        self.gamma           = gamma
        self.label_smoothing = label_smoothing
        self.reduction       = reduction

    def forward(self, logits, targets):
        # Per-sample CE (no reduction) with optional label smoothing
        ce = F.cross_entropy(
            logits, targets,
            weight          = self.weight,
            label_smoothing = self.label_smoothing,
            reduction       = 'none')
        # p_t: probability assigned to the correct class
        pt           = torch.exp(-ce)
        focal_weight = (1.0 - pt) ** self.gamma
        loss         = focal_weight * ce
        return loss.mean() if self.reduction == 'mean' \
               else loss.sum()


# ================================================================
# MODEL
# ================================================================
class ClinicalExcitationNetwork(nn.Module):
    def __init__(self, input_dim, embedding_dim=64,
                 hidden_dim=64, num_classes=3, dropout=0.2):
        super().__init__()

        # SE-style channel excitation gate
        self.excitation = nn.Sequential(
            nn.Linear(input_dim, input_dim // 2),
            nn.ReLU(),
            nn.Linear(input_dim // 2, input_dim),
            nn.Sigmoid()
        )
        # Embedding projection
        self.embedding = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, embedding_dim),
            nn.BatchNorm1d(embedding_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
        )
        # Classification head
        self.classifier = nn.Sequential(
            nn.Linear(embedding_dim, 32),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(32, num_classes)
        )

    def forward(self, x):
        weights    = self.excitation(x)
        x_weighted = x * weights
        embedding  = self.embedding(x_weighted)
        logits     = self.classifier(embedding)
        return logits, embedding, weights


# ================================================================
# TRAINING UTILITIES
# ================================================================
def train_one_epoch(model, loader, optimizer, criterion, device):
    model.train()
    total_loss, correct, total = 0, 0, 0
    for X_batch, y_batch in loader:
        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)
        optimizer.zero_grad()
        logits, _, _ = model(X_batch)
        loss = criterion(logits, y_batch)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * len(y_batch)
        preds       = logits.argmax(dim=1)
        correct    += (preds == y_batch).sum().item()
        total      += len(y_batch)
    return total_loss / total, correct / total


def evaluate(model, loader, criterion, device):
    model.eval()
    total_loss, correct, total = 0, 0, 0
    all_preds, all_labels = [], []
    with torch.no_grad():
        for X_batch, y_batch in loader:
            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)
            logits, _, _ = model(X_batch)
            loss = criterion(logits, y_batch)
            total_loss += loss.item() * len(y_batch)
            preds       = logits.argmax(dim=1)
            correct    += (preds == y_batch).sum().item()
            total      += len(y_batch)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(y_batch.cpu().numpy())
    avg_loss = total_loss / total
    acc      = correct / total
    f1       = f1_score(all_labels, all_preds,
                        average='macro', zero_division=0)
    return avg_loss, acc, f1, all_preds, all_labels


# ================================================================
# HYPERPARAMETERS
# ================================================================
device = torch.device('cuda' if torch.cuda.is_available()
                      else 'cpu')
print(f"Device: {device}")

# Build X and y
X      = df_eng[all_features].copy().astype(float)
y      = df_eng['stage_merged'].values
groups = df_eng['PATNO'].values

INPUT_DIM     = len(all_features)
EMBEDDING_DIM = 64
HIDDEN_DIM    = 64
NUM_CLASSES   = 3          # ← FIXED: was 4, must match label count
DROPOUT       = 0.4        # ← increased from 0.2 (reduces Moderate overfit)
WEIGHT_DECAY  = 5e-3       # ← increased from 1e-3
LR            = 1e-3       # ← restored to 1e-3 (was 3e-4)
EPOCHS        = 200        # ← increased from 100 (focal converges slower)
BATCH_SIZE    = 32
PATIENCE      = 25         # ← increased from 20

FOCAL_GAMMA     = 1.5      # ← standard starting point; try 1.5 or 3.0 if needed
LABEL_SMOOTHING = 0.05     # ← mild smoothing to reduce Moderate overconfidence

labels_map = {0: 'HC', 1: 'Early', 2: 'Moderate'}

print(f"Input dim:         {INPUT_DIM}")
print(f"Embedding:         {EMBEDDING_DIM}")
print(f"Hidden:            {HIDDEN_DIM}")
print(f"Dropout:           {DROPOUT}")
print(f"LR:                {LR}")
print(f"Weight decay:      {WEIGHT_DECAY}")
print(f"Batch size:        {BATCH_SIZE}")
print(f"Patience:          {PATIENCE}")
print(f"Focal gamma:       {FOCAL_GAMMA}")
print(f"Label smoothing:   {LABEL_SMOOTHING}")

print(f"\ny distribution:")
for k, v in zip(*np.unique(y, return_counts=True)):
    print(f"  {k} ({labels_map[k]}): {v}")


# ================================================================
# INITIALIZE OOF EMBEDDINGS + RESULTS STORE
# ================================================================
oof_embeddings = np.zeros((len(X), EMBEDDING_DIM))

sgkf         = StratifiedGroupKFold(
    n_splits=5, shuffle=True, random_state=42)
fold_results = []
cont_idx     = [all_features.index(c)
                for c in continuous_features]


# ================================================================
# 5-FOLD CROSS-VALIDATION LOOP
# ================================================================
for fold, (train_idx, val_idx) in enumerate(
        sgkf.split(X, y, groups=groups)):

    print(f"\n{'='*55}")
    print(f"FOLD {fold+1}/5")
    print(f"{'='*55}")

    # ── Split ─────────────────────────────────────────────────
    X_train_np = X.iloc[train_idx].values.copy()
    X_val_np   = X.iloc[val_idx].values.copy()
    y_train_np = y[train_idx]
    y_val_np   = y[val_idx]

    # ── Scale continuous only (fit on train) ──────────────────
    scaler = StandardScaler()
    X_train_np[:, cont_idx] = scaler.fit_transform(
        X_train_np[:, cont_idx])
    X_val_np[:, cont_idx]   = scaler.transform(
        X_val_np[:, cont_idx])

    # ── Smoothed sqrt class weights ────────────────────────────
    # 1/sqrt(count) → normalize min to 1.0
    # Gentle frequency correction; focal loss handles difficulty
    class_counts = np.bincount(y_train_np)
    smoothed_cw  = 1.0 / np.sqrt(class_counts)
    smoothed_cw  = smoothed_cw / np.min(smoothed_cw)

    cw_tensor = torch.tensor(
        smoothed_cw, dtype=torch.float32).to(device)

    # ── Focal Loss — replaces CrossEntropyLoss ─────────────────
    criterion = FocalLoss(
        weight          = cw_tensor,
        gamma           = FOCAL_GAMMA,
        label_smoothing = LABEL_SMOOTHING)

    print(f"  Class weights (sqrt smoothed):")
    for i, (w, n) in enumerate(zip(smoothed_cw, class_counts)):
        print(f"    {labels_map[i]:<10} "
              f"n={n:3d}  weight={w:.3f}")

    # ── Tensors ───────────────────────────────────────────────
    X_tr = torch.tensor(X_train_np, dtype=torch.float32)
    y_tr = torch.tensor(y_train_np, dtype=torch.long)
    X_vl = torch.tensor(X_val_np,   dtype=torch.float32)
    y_vl = torch.tensor(y_val_np,   dtype=torch.long)

    # ── DataLoaders — shuffle train, fixed val ────────────────
    train_loader = DataLoader(
        TensorDataset(X_tr, y_tr),
        batch_size = BATCH_SIZE,
        shuffle    = True)
    val_loader = DataLoader(
        TensorDataset(X_vl, y_vl),
        batch_size = BATCH_SIZE,
        shuffle    = False)

    # ── Model ─────────────────────────────────────────────────
    model = ClinicalExcitationNetwork(
        input_dim     = INPUT_DIM,
        embedding_dim = EMBEDDING_DIM,
        hidden_dim    = HIDDEN_DIM,
        num_classes   = NUM_CLASSES,
        dropout       = DROPOUT,
    ).to(device)

    optimizer = optim.Adam(
        model.parameters(),
        lr           = LR,
        weight_decay = WEIGHT_DECAY)

    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode     = 'max',   # monitor F1 — higher is better
        patience = 10,
        factor   = 0.5)

    # ── Training loop with early stopping ─────────────────────
    best_val_f1  = 0.0
    best_epoch   = 0
    patience_cnt = 0
    best_state   = None
    history      = []

    for epoch in range(EPOCHS):
        tr_loss, tr_acc = train_one_epoch(
            model, train_loader,
            optimizer, criterion, device)
        vl_loss, vl_acc, vl_f1, _, _ = evaluate(
            model, val_loader, criterion, device)

        scheduler.step(vl_f1)

        history.append({
            'epoch':   epoch + 1,
            'tr_loss': tr_loss, 'tr_acc': tr_acc,
            'vl_loss': vl_loss, 'vl_acc': vl_acc,
            'vl_f1':   vl_f1,
        })

        if vl_f1 > best_val_f1:
            best_val_f1  = vl_f1
            best_epoch   = epoch + 1
            patience_cnt = 0
            best_state   = {
                k: v.clone()
                for k, v in model.state_dict().items()}
        else:
            patience_cnt += 1

        if patience_cnt >= PATIENCE:
            print(f"  Early stop ep={epoch+1} "
                  f"best={best_epoch}")
            break

        if (epoch + 1) % 10 == 0:
            print(f"  Ep {epoch+1:3d} | "
                  f"tr={tr_loss:.4f}/{tr_acc:.3f} | "
                  f"vl={vl_loss:.4f}/{vl_acc:.3f} "
                  f"f1={vl_f1:.3f}")

    # ── Load best weights ─────────────────────────────────────
    model.load_state_dict(best_state)
    _, val_acc, val_f1, preds, labels_true = evaluate(
        model, val_loader, criterion, device)

    # ── Extract OOF embeddings (no leakage) ───────────────────
    # Val patients were never seen during this fold's training
    model.eval()
    with torch.no_grad():
        _, fold_embs, _ = model(X_vl.to(device))
        oof_embeddings[val_idx] = fold_embs.cpu().numpy()

    print(f"\n  Best epoch:   {best_epoch}")
    print(f"  Val Acc:      {val_acc:.4f}")
    print(f"  Val F1 macro: {val_f1:.4f}")
    print(f"\n  Classification Report:")
    print(classification_report(
        labels_true, preds,
        target_names=list(labels_map.values()),
        zero_division=0))

    print(f"  Val class distribution:")
    for k, v in zip(*np.unique(y_val_np, return_counts=True)):
        print(f"    {labels_map[k]:<10}: {v}")

    fold_results.append({
        'fold':        fold + 1,
        'val_acc':     val_acc,
        'val_f1':      val_f1,
        'best_epoch':  best_epoch,
        'preds':       preds,
        'labels':      labels_true,
        'history':     history,
        'scaler':      scaler,
        'best_state':  best_state,
        'smoothed_cw': smoothed_cw,
    })


# ================================================================
# FINAL SUMMARY
# ================================================================
print(f"\n{'='*55}")
print("5-FOLD CROSS-VALIDATION SUMMARY")
print(f"{'='*55}")
accs = [r['val_acc']    for r in fold_results]
f1s  = [r['val_f1']     for r in fold_results]
eps  = [r['best_epoch'] for r in fold_results]

for r in fold_results:
    print(f"  Fold {r['fold']}: "
          f"Acc={r['val_acc']:.4f}  "
          f"F1={r['val_f1']:.4f}  "
          f"Best_ep={r['best_epoch']}")

print(f"\n  Mean Acc: {np.mean(accs):.4f} "
      f"± {np.std(accs):.4f}")
print(f"  Mean F1:  {np.mean(f1s):.4f} "
      f"± {np.std(f1s):.4f}")

avg_best_epoch = int(np.mean(eps))
print(f"  Avg best epoch: {avg_best_epoch}")


# ================================================================
# SAVE BEST FOLD MODEL
# ================================================================
best_fold_idx = int(np.argmax(f1s))
best_fold     = fold_results[best_fold_idx]
best_state    = best_fold['best_state']
best_scaler   = best_fold['scaler']

print(f"\nBest fold: {best_fold['fold']} "
      f"(F1={best_fold['val_f1']:.4f})")

# Class weights on full dataset for inference-time use
final_counts = np.bincount(y)
final_cw     = 1.0 / np.sqrt(final_counts)
final_cw     = final_cw / np.min(final_cw)

print(f"\nFull dataset class weights:")
for i, (w, n) in enumerate(zip(final_cw, final_counts)):
    print(f"  {labels_map[i]:<10} n={n:3d}  weight={w:.3f}")

torch.save({
    'model_state':         best_state,
    'input_dim':           INPUT_DIM,
    'embedding_dim':       EMBEDDING_DIM,
    'hidden_dim':          HIDDEN_DIM,
    'num_classes':         NUM_CLASSES,
    'dropout':             DROPOUT,
    'continuous_features': continuous_features,
    'binary_features':     binary_features,
    'all_features':        all_features,
    'labels_map':          labels_map,
    'class_weights':       final_cw.tolist(),
    'cont_idx':            cont_idx,
    'focal_gamma':         FOCAL_GAMMA,         # ← saved for reproducibility
    'label_smoothing':     LABEL_SMOOTHING,     # ← saved for reproducibility
    'cv_mean_f1':          float(np.mean(f1s)),
    'cv_std_f1':           float(np.std(f1s)),
    'best_fold':           best_fold['fold'],
    'avg_best_epoch':      avg_best_epoch,
}, 'clinical_excitation_model.pt')

joblib.dump(best_scaler, 'clinical_scaler.pkl')


# ================================================================
# SAVE OOF EMBEDDINGS WITH PATIENT IDs
# ================================================================
zeros_remaining = (oof_embeddings.sum(axis=1) == 0).sum()
print(f"\nOOF zeros remaining: {zeros_remaining} ← must be 0")

ref_cols = [
    'PATNO', 'EVENT_ID',
    'Image_ID_T1', 'Image_ID_T2',
    'NSD_ISS_Stage', 'label_binary', 'stage_merged'
]
ref_cols = [c for c in ref_cols if c in df_eng.columns]
ref_df   = df_eng[ref_cols].reset_index(drop=True)

emb_cols = [f'emb_{i}' for i in range(EMBEDDING_DIM)]
emb_df   = pd.DataFrame(oof_embeddings, columns=emb_cols)
final_df = pd.concat([ref_df, emb_df], axis=1)

final_df.to_csv('clinical_embeddings_oof.csv', index=False)
np.save('clinical_embeddings_oof.npy', oof_embeddings)

print(f"\n✅ Saved: clinical_excitation_model.pt")
print(f"✅ Saved: clinical_scaler.pkl")
print(f"✅ Saved: clinical_embeddings_oof.csv "
      f"({final_df.shape})")
print(f"✅ Saved: clinical_embeddings_oof.npy "
      f"({oof_embeddings.shape})")


# ================================================================
# VERIFICATION
# ================================================================
print(f"\n{'='*55}")
print("VERIFICATION")
print(f"{'='*55}")
print(f"  Unique patients:     {final_df['PATNO'].nunique()}")
print(f"  Total visits:        {len(final_df)}")
print(f"  T1 images linked:    "
      f"{final_df['Image_ID_T1'].notna().sum()}")
print(f"  T2 images linked:    "
      f"{final_df['Image_ID_T2'].notna().sum()}")
print(f"  Embedding NaN:       "
      f"{np.isnan(oof_embeddings).any()}")
print(f"  Embedding shape:     {oof_embeddings.shape}")
print(f"  CV Mean F1:          {np.mean(f1s):.4f} "
      f"± {np.std(f1s):.4f}")

print(f"\n  Stage distribution in embeddings:")
for k, v in zip(*np.unique(
        final_df['stage_merged'], return_counts=True)):
    print(f"    {labels_map[k]:<10}: {v}")


# ================================================================
# LEAKAGE CHECK
# ================================================================
print(f"\n{'='*55}")
print("HC vs PD leakage check")
print(f"{'='*55}")
check_cols = [c for c in [
    'Part1', 'Part1P', 'RBD_score'
] if c in X.columns]

print(f"\n{'Feature':<22} {'HC mean':>10} "
      f"{'PD mean':>10} {'Flag':>10}")
print("-" * 57)
for col in check_cols:
    hc_mean = X[y == 0][col].mean()
    pd_mean = X[y != 0][col].mean()
    ratio   = abs(pd_mean - hc_mean) / (abs(hc_mean) + 1e-6)
    flag    = "🔴 LEAKY" if ratio > 5 else (
              "⚠️  high"  if ratio > 2 else "✅ ok")
    print(f"  {col:<20} {hc_mean:>10.3f} "
          f"{pd_mean:>10.3f} {flag:>10}")
```
